```python
import pandas as pd

def get_match_tweets(df, match_name):
    match = MATCHES[match_name]
    # Convertimos la fecha del diccionario a objeto date de python
    match_date = pd.to_datetime(match["date"]).date()

    team1_keywords = match["keywords1"]
    team2_keywords = match["keywords2"]

    # Filtramos por fecha (asegurándonos que la columna Date sea datetime)
    df['Date'] = pd.to_datetime(df['Date'])
    df_day = df[df["Date"].dt.date == match_date].copy()

    # Regex para keywords
    team1_regex = "|".join(team1_keywords)
    team2_regex = "|".join(team2_keywords)

    # Buscamos menciones en la columna 'Tweet'
    mask = (
        df_day["Tweet"].str.contains(
            team1_regex + "|" + team2_regex,
            case=False,
            na=False
        )
    )

    return df_day[mask].copy()

# Ejecutamos para la final
try:
    tweets = get_match_tweets(df, "France vs Croatia")
    print(f"Tweets encontrados para la final: {len(tweets)}")
    display(tweets.head())
except NameError:
    print("Error: La variable 'df' no existe. Ejecuta primero la celda de carga de datos.")
```

```python
def clean_match_tweets(df_match):

    df_match = df_match.copy()

    df_match = df_match.dropna(subset=["Tweet"])

    df_match = df_match.drop_duplicates(subset=["Tweet"])

    return df_match
```

```python
# =========================================================
# ETAPAS VISUALES DEL PARTIDO
# =========================================================

def build_period_layout(match_name):

    if match_name not in MATCH_PERIODS:
        raise ValueError(
            f"No existe configuración temporal para: {match_name}"
        )

    configuration = MATCH_PERIODS[match_name]

    periods = [
        {
            "key": "first_half",
            "label": "PRIMER TIEMPO",
            "type": "play",
            "football_start": 0,
            "football_end": 45,
            "display_width": 45
        },
        {
            "key": "halftime",
            "label": "ENTRETIEMPO",
            "type": "break",
            "football_start": None,
            "football_end": None,
            "display_width": 10
        },
        {
            "key": "second_half",
            "label": "SEGUNDO TIEMPO",
            "type": "play",
            "football_start": 45,
            "football_end": 90,
            "display_width": 45
        }
    ]

    if configuration["has_extra_time"]:

        periods.extend([
            {
                "key": "extra_time_transition",
                "label": "TIEMPO SUPLEMENTARIO",
                "type": "transition",
                "football_start": None,
                "football_end": None,
                "display_width": 14
            },
            {
                "key": "extra_time_first_half",
                "label": "PRIMER TIEMPO SUPLEMENTARIO",
                "type": "play",
                "football_start": 90,
                "football_end": 105,
                "display_width": 15
            },
            {
                "key": "extra_time_halftime",
                "label": "ENTRETIEMPO",
                "type": "break",
                "football_start": None,
                "football_end": None,
                "display_width": 10
            },
            {
                "key": "extra_time_second_half",
                "label": "SEGUNDO TIEMPO SUPLEMENTARIO",
                "type": "play",
                "football_start": 105,
                "football_end": 120,
                "display_width": 15
            }
        ])

    if configuration["has_shootout"]:

        periods.append({
            "key": "shootout",
            "label": "PENALES",
            "type": "shootout",
            "football_start": None,
            "football_end": None,
            "display_width": 24
        })

    # Posición inicial y final de cada etapa.
    cursor = 0

    for period in periods:

        period["display_start"] = cursor

        cursor += period["display_width"]

        period["display_end"] = cursor

    return periods
```

```python
# =========================================================
# DURACIÓN REAL APROXIMADA DE CADA ETAPA
# =========================================================

PERIOD_REAL_DURATIONS = {
    "first_half": 45,
    "halftime": 15,
    "second_half": 45,
    "extra_time_transition": 5,
    "extra_time_first_half": 15,
    "extra_time_halftime": 2,
    "extra_time_second_half": 15,
    "shootout": 20
}


def elapsed_to_match_display(
    elapsed_minutes,
    match_name
):
    """
    Convierte los minutos transcurridos desde el kickoff
    en una posición interna del gráfico.
    """

    is_scalar = np.isscalar(elapsed_minutes)

    elapsed = np.atleast_1d(
        np.asarray(elapsed_minutes, dtype=float)
    )

    display = np.full(
        elapsed.shape,
        np.nan,
        dtype=float
    )

    periods = build_period_layout(match_name)

    real_cursor = 0

    for index, period in enumerate(periods):

        real_duration = PERIOD_REAL_DURATIONS[
            period["key"]
        ]

        real_start = real_cursor
        real_end = real_start + real_duration

        if index == len(periods) - 1:

            mask = (
                (elapsed >= real_start)
                & (elapsed <= real_end)
            )

        else:

            mask = (
                (elapsed >= real_start)
                & (elapsed < real_end)
            )

        progress = (
            (elapsed[mask] - real_start)
            / real_duration
        )

        display[mask] = (
            period["display_start"]
            + progress * period["display_width"]
        )

        real_cursor = real_end

    if is_scalar:
        return float(display[0])

    return display


def football_minute_to_match_display(
    football_minute,
    match_name
):
    """
    Coloca un gol o evento según su minuto futbolístico.
    """

    minute = float(football_minute)

    play_periods = [
        period
        for period in build_period_layout(match_name)
        if period["type"] == "play"
    ]

    for period in play_periods:

        start = period["football_start"]
        end = period["football_end"]

        if start <= minute <= end:

            progress = (
                (minute - start)
                / (end - start)
            )

            return (
                period["display_start"]
                + progress * period["display_width"]
            )

    return np.nan


def shootout_order_to_match_display(
    penalty_order,
    match_name
):
    """
    Distribuye los penales dentro de la franja PENALES.
    """

    shootout_period = next(
        (
            period
            for period in build_period_layout(match_name)
            if period["key"] == "shootout"
        ),
        None
    )

    if shootout_period is None:
        return np.nan

    penalties = MATCH_SHOOTOUTS.get(
        match_name,
        []
    )

    total_penalties = len(penalties)

    return (
        shootout_period["display_start"]
        + (
            penalty_order
            / (total_penalties + 1)
        )
        * shootout_period["display_width"]
    )
```

```python
# =========================================================
# FUNCIONES AUXILIARES
# =========================================================

def build_keyword_pattern(keywords):

    escaped_keywords = [
        re.escape(str(keyword).strip())
        for keyword in keywords
        if str(keyword).strip()
    ]

    joined_keywords = "|".join(
        escaped_keywords
    )

    return (
        rf"(?<!\w)"
        rf"(?:{joined_keywords})"
        rf"(?!\w)"
    )


def hex_to_rgba(hex_color, opacity):

    color = hex_color.lstrip("#")

    red = int(color[0:2], 16)
    green = int(color[2:4], 16)
    blue = int(color[4:6], 16)

    return (
        f"rgba("
        f"{red}, {green}, {blue}, {opacity}"
        f")"
    )


def get_visible_line_color(color):

    if color.upper() == "#171717":
        return "#A3A3A3"

    if color.upper() == "#F4F4F4":
        return "#FFFFFF"

    return color


def get_event_team(
    match_name,
    event_index,
    event
):

    # En el futuro los eventos tendrán el equipo incorporado.
    # Si ya existe, lo usamos.

    if event.get("team"):
        return event["team"]

    teams = EVENT_TEAM_ORDER.get(
        match_name,
        []
    )

    if event_index < len(teams):
        return teams[event_index]

    return None


def get_event_type(
    match_name,
    event
):

    if event.get("type"):
        return event["type"]

    event_key = (
        match_name,
        int(event["minute"])
    )

    if event_key in EVENT_TYPE_OVERRIDES:
        return EVENT_TYPE_OVERRIDES[event_key]

    event_text = event.get(
        "event",
        ""
    ).lower()

    if "ataja" in event_text:
        return "save"

    if (
        "penales" in event_text
        or "tanda" in event_text
        or "clasifica" in event_text
    ):
        return "shootout"

    return "goal"
```

```python
import os
import glob
import base64
import unicodedata
from functools import lru_cache


# =========================================================
# BANDERAS
# =========================================================

TEAM_FLAG_NAMES = {
    "France": "Francia",
    "Argentina": "Argentina",
    "Uruguay": "Uruguay",
    "Portugal": "Portugal",
    "Brazil": "Brasil",
    "Mexico": "México",
    "Belgium": "Bélgica",
    "Japan": "Japón",
    "Spain": "España",
    "Russia": "Rusia",
    "Croatia": "Croacia",
    "Denmark": "Dinamarca",
    "Sweden": "Suecia",
    "Switzerland": "Suiza",
    "Colombia": "Colombia",
    "England": "Inglaterra"
}


def normalize_flag_name(name):

    name = unicodedata.normalize(
        "NFD",
        str(name)
    )

    name = "".join(
        character
        for character in name
        if unicodedata.category(character) != "Mn"
    )

    return name.lower().strip()


@lru_cache(maxsize=32)
def load_flag_data_uri(team_name):

    expected_name = normalize_flag_name(
        TEAM_FLAG_NAMES.get(
            team_name,
            team_name
        )
    )

    patterns = [
        "/content/drive/MyDrive/**/países/**/*.*",
        "/content/drive/MyDrive/**/paises/**/*.*"
    ]

    for pattern in patterns:

        for file_path in glob.glob(
            pattern,
            recursive=True
        ):

            if not os.path.isfile(file_path):
                continue

            file_name = os.path.splitext(
                os.path.basename(file_path)
            )[0]

            if (
                normalize_flag_name(file_name)
                != expected_name
            ):
                continue

            extension = (
                file_path
                .lower()
                .split(".")[-1]
            )

            if extension == "jpg":
                extension = "jpeg"

            if extension not in [
                "png",
                "jpeg",
                "webp"
            ]:
                continue

            with open(
                file_path,
                "rb"
            ) as image_file:

                encoded_image = (
                    base64
                    .b64encode(
                        image_file.read()
                    )
                    .decode()
                )

            return (
                f"data:image/{extension};"
                f"base64,{encoded_image}"
            )

    return None
```

```python
# =========================================================
# RANGOS SIN COBERTURA DEL DATASET
# =========================================================

def get_missing_coverage_ranges(momentum):
    """
    Agrupa intervalos consecutivos sin cobertura para
    representarlos como una única zona en el gráfico.
    """

    missing = momentum.loc[
        ~momentum["has_coverage"],
        [
            "display_start",
            "display_end"
        ]
    ]

    if missing.empty:
        return []

    ranges = []

    current_start = float(
        missing.iloc[0]["display_start"]
    )

    current_end = float(
        missing.iloc[0]["display_end"]
    )

    for _, row in missing.iloc[1:].iterrows():

        start = float(
            row["display_start"]
        )

        end = float(
            row["display_end"]
        )

        if np.isclose(
            start,
            current_end
        ):
            current_end = end

        else:
            ranges.append(
                (
                    current_start,
                    current_end
                )
            )

            current_start = start
            current_end = end

    ranges.append(
        (
            current_start,
            current_end
        )
    )

    return ranges
```

```python
# PULSO DEL PARTIDO — VERSIÓN FINAL POR ETAPAS
# =========================================================
# ETAPAS CON TIEMPO REAL Y POSICIÓN VISUAL
# =========================================================

def get_runtime_periods(match_name):

    periods = build_period_layout(match_name)

    real_cursor = 0
    runtime_periods = []

    for period in periods:

        period_data = period.copy()

        real_duration = PERIOD_REAL_DURATIONS[
            period["key"]
        ]

        period_data["real_start"] = real_cursor
        period_data["real_end"] = (
            real_cursor + real_duration
        )

        runtime_periods.append(period_data)

        real_cursor += real_duration

    return runtime_periods


# =========================================================
# DATOS DEL PULSO POR ETAPAS
# =========================================================

def build_twitter_momentum_final(
    df,
    match_name,
    interval_minutes=5
):

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    if match_name not in MATCH_KICKOFFS:
        raise ValueError(
            f"No existe kickoff para: {match_name}"
        )

    if match_name not in MATCH_PERIODS:
        raise ValueError(
            f"No existe formato para: {match_name}"
        )

    source = df.copy()

    source["Date"] = pd.to_datetime(
        source["Date"],
        errors="coerce"
    )

    source = source.dropna(
        subset=["Date"]
    )

    match = MATCHES[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    kickoff = pd.to_datetime(
        MATCH_KICKOFFS[match_name]
    )

    match_date = pd.to_datetime(
        match["date"]
    ).date()

    day_tweets = source[
        source["Date"].dt.date == match_date
    ].copy()

    day_tweets["elapsed_minute"] = (
        day_tweets["Date"] - kickoff
    ).dt.total_seconds() / 60

    runtime_periods = get_runtime_periods(
        match_name
    )

    real_match_end = runtime_periods[-1][
        "real_end"
    ]

    day_tweets = day_tweets[
        day_tweets["elapsed_minute"].between(
            0,
            real_match_end,
            inclusive="both"
        )
    ].copy()

    team1_pattern = build_keyword_pattern(
        match["keywords1"]
    )

    team2_pattern = build_keyword_pattern(
        match["keywords2"]
    )

    tweet_text = day_tweets[
        "Tweet"
    ].fillna("")

    team1_mask = tweet_text.str.contains(
        team1_pattern,
        case=False,
        regex=True,
        na=False
    )

    team2_mask = tweet_text.str.contains(
        team2_pattern,
        case=False,
        regex=True,
        na=False
    )

    rows = []

    for period in runtime_periods:

        real_start = period["real_start"]
        real_end = period["real_end"]

        real_duration = real_end - real_start

        bin_start = real_start

        while bin_start < real_end:

            bin_end = min(
                bin_start + interval_minutes,
                real_end
            )

            bin_mask = (
                (
                    day_tweets["elapsed_minute"]
                    >= bin_start
                )
                &
                (
                    day_tweets["elapsed_minute"]
                    < bin_end
                )
            )

            coverage_count = int(
                bin_mask.sum()
            )

            team1_count = int(
                (bin_mask & team1_mask).sum()
            )

            team2_count = int(
                (bin_mask & team2_mask).sum()
            )

            start_progress = (
                (bin_start - real_start)
                / real_duration
            )

            end_progress = (
                (bin_end - real_start)
                / real_duration
            )

            display_start = (
                period["display_start"]
                + start_progress
                * period["display_width"]
            )

            display_end = (
                period["display_start"]
                + end_progress
                * period["display_width"]
            )

            display_middle = (
                display_start + display_end
            ) / 2

            if period["type"] == "play":

                football_duration = (
                    period["football_end"]
                    - period["football_start"]
                )

                football_start = (
                    period["football_start"]
                    + start_progress
                    * football_duration
                )

                football_end = (
                    period["football_start"]
                    + end_progress
                    * football_duration
                )

                period_label = (
                    f"{period['label'].title()} · "
                    f"{football_start:.0f}′–"
                    f"{football_end:.0f}′"
                )

            else:

                period_label = period[
                    "label"
                ].title()

            has_coverage = (
                coverage_count > 0
            )

            rows.append({
                "minute": display_middle,
                "display_start": display_start,
                "display_end": display_end,
                "elapsed_start": bin_start,
                "elapsed_end": bin_end,
                "period_key": period["key"],
                "period_label": period_label,

                "team1_mentions": (
                    float(team1_count)
                    if has_coverage
                    else np.nan
                ),

                "team2_mentions": (
                    float(team2_count)
                    if has_coverage
                    else np.nan
                ),

                "team2_plot": (
                    -float(team2_count)
                    if has_coverage
                    else np.nan
                ),

                "has_coverage": has_coverage
            })

            bin_start = bin_end

    momentum = pd.DataFrame(rows)

    metadata = {
        "match": match_name,
        "team1": team1,
        "team2": team2,
        "interval_minutes": interval_minutes,
        "periods": runtime_periods,
        "display_end": runtime_periods[-1][
            "display_end"
        ],
        "has_extra_time": MATCH_PERIODS[
            match_name
        ]["has_extra_time"],
        "has_shootout": MATCH_PERIODS[
            match_name
        ]["has_shootout"],
        "events": MATCH_EVENTS[
            match_name
        ]["events"],
        "shootout": MATCH_SHOOTOUTS.get(
            match_name,
            []
        )
    }

    return momentum, metadata
```

```python
# =========================================================
# GRÁFICO FINAL DEL PULSO DEL PARTIDO
# =========================================================

def build_final_axis(metadata):
    """
    Construye las etiquetas visibles del eje.
    Los descansos y los penales se muestran con palabras, no con minutos.
    """

    tick_values = []
    tick_text = []
    guide_values = []

    for period in metadata["periods"]:

        start_x = period["display_start"]
        end_x = period["display_end"]

        if period["type"] == "play":

            football_start = period["football_start"]
            football_end = period["football_end"]
            football_duration = football_end - football_start

            if football_duration == 45:
                football_ticks = list(
                    np.arange(
                        football_start,
                        football_end + 0.1,
                        15
                    )
                )
            else:
                football_ticks = [
                    football_start,
                    football_end
                ]

            for football_minute in football_ticks:

                progress = (
                    (football_minute - football_start)
                    / football_duration
                )

                display_x = (
                    start_x
                    + progress
                    * period["display_width"]
                )

                tick_values.append(display_x)
                tick_text.append(
                    f"{int(football_minute)}′"
                )
                guide_values.append(display_x)

        else:

            center_x = (start_x + end_x) / 2

            if period["key"] == "extra_time_transition":
                label = (
                    "<b>TIEMPO<br>"
                    "SUPLEMENTARIO</b>"
                )

            else:
                label = (
                    f"<b>{period['label']}</b>"
                )

            tick_values.append(center_x)
            tick_text.append(label)

    return tick_values, tick_text, guide_values


def plot_twitter_momentum_final(
    df,
    match_name,
    interval_minutes=5
):

    momentum, metadata = (
        build_twitter_momentum_final(
            df=df,
            match_name=match_name,
            interval_minutes=interval_minutes
        )
    )

    team1 = metadata["team1"]
    team2 = metadata["team2"]

    colors = TEAM_COLORS.copy()

    colors.update(
        MATCH_COLOR_OVERRIDES.get(
            match_name,
            {}
        )
    )

    team1_color = colors.get(
        team1,
        "#2457D6"
    )

    team2_color = colors.get(
        team2,
        "#75AADB"
    )

    team1_line = get_visible_line_color(
        team1_color
    )

    team2_line = get_visible_line_color(
        team2_color
    )

    available_mentions = pd.concat([
        momentum["team1_mentions"],
        momentum["team2_mentions"]
    ]).dropna()

    if (
        available_mentions.empty
        or available_mentions.max() <= 0
    ):
        max_mentions = 1
    else:
        max_mentions = float(
            available_mentions.max()
        )

    vertical_limit = max_mentions * 1.30

    figure = go.Figure()

    # =====================================================
    # CURVA DEL EQUIPO SUPERIOR
    # =====================================================

    team1_hover = np.column_stack([
        momentum["period_label"],
        momentum["team1_mentions"]
    ])

    figure.add_trace(
        go.Scatter(
            x=momentum["minute"],
            y=momentum["team1_mentions"],
            mode="lines",
            line={
                "color": team1_line,
                "width": 3,
                "shape": "spline",
                "smoothing": 1.10
            },
            fill="tozeroy",
            fillcolor=hex_to_rgba(
                team1_color,
                0.76
            ),
            connectgaps=False,
            name=team1,
            customdata=team1_hover,
            hovertemplate=(
                f"<b>{team1}</b><br>"
                "%{customdata[0]}<br>"
                "%{customdata[1]:.0f} menciones"
                "<extra></extra>"
            )
        )
    )

    # =====================================================
    # CURVA INVERTIDA DEL EQUIPO INFERIOR
    # =====================================================

    team2_hover = np.column_stack([
        momentum["period_label"],
        momentum["team2_mentions"]
    ])

    figure.add_trace(
        go.Scatter(
            x=momentum["minute"],
            y=momentum["team2_plot"],
            mode="lines",
            line={
                "color": team2_line,
                "width": 3,
                "shape": "spline",
                "smoothing": 1.10
            },
            fill="tozeroy",
            fillcolor=hex_to_rgba(
                team2_color,
                0.76
            ),
            connectgaps=False,
            name=team2,
            customdata=team2_hover,
            hovertemplate=(
                f"<b>{team2}</b><br>"
                "%{customdata[0]}<br>"
                "%{customdata[1]:.0f} menciones"
                "<extra></extra>"
            )
        )
    )

    # =====================================================
    # FRANJAS DE DESCANSO, ALARGUE Y PENALES
    # =====================================================

    period_colors = {
        "halftime": "rgba(212,175,55,0.12)",
        "extra_time_transition":
            "rgba(29,78,158,0.17)",
        "extra_time_halftime":
            "rgba(212,175,55,0.12)",
        "shootout": "rgba(213,43,30,0.13)"
    }

    for period in metadata["periods"]:

        if period["type"] == "play":
            continue

        figure.add_vrect(
            x0=period["display_start"],
            x1=period["display_end"],
            fillcolor=period_colors.get(
                period["key"],
                "rgba(255,255,255,0.08)"
            ),
            line_width=0,
            layer="below"
        )

        figure.add_vline(
            x=period["display_start"],
            line_color=(
                "rgba(255,255,255,0.38)"
            ),
            line_width=1
        )

        figure.add_vline(
            x=period["display_end"],
            line_color=(
                "rgba(255,255,255,0.38)"
            ),
            line_width=1
        )

    # =====================================================
    # ZONAS SIN COBERTURA
    # =====================================================

    missing_ranges = (
        get_missing_coverage_ranges(
            momentum
        )
    )

    for start, end in missing_ranges:

        figure.add_vrect(
            x0=start,
            x1=end,
            fillcolor="rgba(0,0,0,0.17)",
            line_width=0,
            layer="below"
        )

        if end - start >= 20:

            figure.add_annotation(
                x=(start + end) / 2,
                y=0,
                text=(
                    "SIN COBERTURA "
                    "DEL DATASET"
                ),
                textangle=-90,
                showarrow=False,
                font={
                    "family":
                        "Source Sans Pro, Arial",
                    "size": 10,
                    "color":
                        "rgba(255,255,255,0.30)"
                }
            )

    # =====================================================
    # EJE CENTRAL
    # =====================================================

    figure.add_hline(
        y=0,
        line_color="#F2F2F2",
        line_width=2
    )

    # =====================================================
    # GOLES Y EVENTOS DEL PARTIDO
    # =====================================================

    icons = {
        "goal": "⚽",
        "save": "🧤"
    }

    event_distance = max_mentions * 0.18

    for event_index, event in enumerate(
        metadata["events"]
    ):

        event_type = get_event_type(
            match_name,
            event
        )

        # La tanda se muestra lanzamiento por lanzamiento.
        if (
            event_type == "shootout"
            and metadata["has_shootout"]
        ):
            continue

        event_minute = int(
            event["minute"]
        )

        event_x = (
            football_minute_to_match_display(
                event_minute,
                match_name
            )
        )

        if pd.isna(event_x):
            continue

        event_team = get_event_team(
            match_name,
            event_index,
            event
        )

        if event_team == team1:
            direction = 1
        elif event_team == team2:
            direction = -1
        else:
            direction = 1

        separation_factor = (
            1
            + 0.16
            * (event_index % 2)
        )

        event_y = (
            direction
            * event_distance
            * separation_factor
        )

        icon = icons.get(
            event_type,
            "●"
        )

        figure.add_annotation(
            x=event_x,
            y=event_y,
            ax=event_x,
            ay=0,
            xref="x",
            yref="y",
            axref="x",
            ayref="y",
            text="",
            showarrow=True,
            arrowhead=2,
            arrowsize=1,
            arrowwidth=2,
            arrowcolor="#F2F2F2"
        )

        figure.add_trace(
            go.Scatter(
                x=[event_x],
                y=[event_y],
                mode="text",
                text=[icon],
                textfont={
                    "size": 17,
                    "color": "#FFFFFF"
                },
                showlegend=False,
                hovertemplate=(
                    f"<b>{event_minute}′</b><br>"
                    f"{event.get('event', '')}"
                    "<extra></extra>"
                )
            )
        )

    # =====================================================
    # TANDA DE PENALES
    # =====================================================

    penalty_result_labels = {
        "scored": "Convertido",
        "saved": "Atajado",
        "missed": "Errado"
    }

    penalty_distance = max_mentions * 0.28

    for penalty in metadata["shootout"]:

        penalty_x = (
            shootout_order_to_match_display(
                penalty["order"],
                match_name
            )
        )

        if penalty["team"] == team1:
            penalty_y = penalty_distance
        else:
            penalty_y = -penalty_distance

        penalty_icon = (
            "⚽"
            if penalty["outcome"] == "scored"
            else "✕"
        )

        figure.add_trace(
            go.Scatter(
                x=[penalty_x],
                y=[penalty_y],
                mode="text",
                text=[penalty_icon],
                textfont={
                    "size": (
                        18
                        if penalty["outcome"]
                        == "scored"
                        else 22
                    ),
                    "color": "#FFFFFF"
                },
                showlegend=False,
                hovertemplate=(
                    f"<b>{penalty['player']}</b><br>"
                    f"{penalty['team']}<br>"
                    f"{penalty_result_labels.get(
                        penalty['outcome'],
                        penalty['outcome']
                    )}"
                    "<extra></extra>"
                )
            )
        )

    # =====================================================
    # EJE TEMPORAL
    # =====================================================

    (
        tick_values,
        tick_text,
        guide_values
    ) = build_final_axis(metadata)

    for guide_x in guide_values:

        figure.add_vline(
            x=guide_x,
            line_color=(
                "rgba(255,255,255,0.15)"
            ),
            line_width=1,
            line_dash="dot",
            layer="below"
        )

    tick_size = (
        10
        if metadata["has_extra_time"]
        else 12
    )

    # =====================================================
    # DISEÑO GENERAL
    # =====================================================

    figure.update_layout(
        title={
            "text": (
                "<b>"
                "PULSO DEL PARTIDO EN TWITTER"
                "</b>"
                "<br>"
                "<span style='font-size:16px;'>"
                "Menciones por equipo "
                "en intervalos de cinco minutos"
                "</span>"
            ),
            "x": 0.5,
            "xanchor": "center",
            "font": {
                "family":
                    "Source Sans Pro, Arial, sans-serif",
                "size": 31,
                "color": "#FFFFFF"
            }
        },
        height=650,
        margin={
            "l": 125,
            "r": 45,
            "t": 145,
            "b": 105
        },
        paper_bgcolor="#050505",
        plot_bgcolor="#3B3F41",
        font={
            "family":
                "Source Sans Pro, Arial, sans-serif",
            "color": "#FFFFFF"
        },
        showlegend=False,
        hovermode="x",
        xaxis={
            "range": [
                0,
                metadata["display_end"]
            ],
            "tickmode": "array",
            "tickvals": tick_values,
            "ticktext": tick_text,
            "tickfont": {
                "size": tick_size,
                "color": "#FFFFFF"
            },
            "showgrid": False,
            "showline": True,
            "linecolor": "#FFFFFF",
            "linewidth": 1,
            "zeroline": False,
            "title": None,
            "automargin": True
        },
        yaxis={
            "range": [
                -vertical_limit,
                vertical_limit
            ],
            "showticklabels": False,
            "showgrid": False,
            "showline": True,
            "linecolor": "#FFFFFF",
            "linewidth": 1,
            "zeroline": False,
            "title": None
        }
    )

    # =====================================================
    # BANDERAS
    # =====================================================

    team1_flag = load_flag_data_uri(
        team1
    )

    team2_flag = load_flag_data_uri(
        team2
    )

    if team1_flag:

        figure.add_layout_image({
            "source": team1_flag,
            "xref": "paper",
            "yref": "paper",
            "x": -0.085,
            "y": 0.76,
            "sizex": 0.075,
            "sizey": 0.075,
            "xanchor": "center",
            "yanchor": "middle",
            "sizing": "contain",
            "layer": "above"
        })

    else:

        figure.add_annotation(
            xref="paper",
            yref="paper",
            x=-0.04,
            y=0.76,
            text=f"<b>{team1}</b>",
            showarrow=False,
            font={
                "color": team1_line,
                "size": 12
            },
            xanchor="right"
        )

    if team2_flag:

        figure.add_layout_image({
            "source": team2_flag,
            "xref": "paper",
            "yref": "paper",
            "x": -0.085,
            "y": 0.24,
            "sizex": 0.075,
            "sizey": 0.075,
            "xanchor": "center",
            "yanchor": "middle",
            "sizing": "contain",
            "layer": "above"
        })

    else:

        figure.add_annotation(
            xref="paper",
            yref="paper",
            x=-0.04,
            y=0.24,
            text=f"<b>{team2}</b>",
            showarrow=False,
            font={
                "color": team2_line,
                "size": 12
            },
            xanchor="right"
        )

    # =====================================================
    # MARCO EXTERIOR
    # =====================================================

    frame_left = -0.17
    frame_right = 1.06
    frame_bottom = -0.27
    frame_top = 1.38

    frame = {
        "xref": "paper",
        "yref": "paper",
        "layer": "above"
    }

    figure.add_shape(
        type="rect",
        x0=frame_left,
        x1=frame_right,
        y0=frame_bottom,
        y1=frame_top,
        line={
            "color": "#D4AF37",
            "width": 6
        },
        fillcolor="rgba(0,0,0,0)",
        **frame
    )

    frame_width = frame_right - frame_left

    for start, end, color in [
        (
            frame_left,
            frame_left + frame_width * 0.34,
            "#D52B1E"
        ),
        (
            frame_left + frame_width * 0.34,
            frame_left + frame_width * 0.68,
            "#1D4E9E"
        ),
        (
            frame_left + frame_width * 0.68,
            frame_right,
            "#D4AF37"
        )
    ]:

        figure.add_shape(
            type="line",
            x0=start,
            x1=end,
            y0=frame_top,
            y1=frame_top,
            line={
                "color": color,
                "width": 12
            },
            **frame
        )

    return figure, momentum


# =========================================================
# PRUEBA: ALARGUE Y PENALES
# =========================================================

fig_final, momentum_final = (
    plot_twitter_momentum_final(
        df=df,
        match_name="Croatia vs Denmark",
        interval_minutes=5
    )
)

fig_final.show(
    config={
        "displayModeBar": False,
        "responsive": True
    }
)
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 1. CONFIGURACIÓN DE FASES
# =========================================================

MATCH_STAGE_LABELS = {
    "round_of_16": "Octavos de final",
    "quarterfinals": "Cuartos de final",
    "semifinals": "Semifinal",
    "third_place": "Tercer puesto",
    "final": "Final"
}


MATCH_STAGES = {
    # Octavos de final
    "France vs Argentina": "round_of_16",
    "Uruguay vs Portugal": "round_of_16",
    "Spain vs Russia": "round_of_16",
    "Croatia vs Denmark": "round_of_16",
    "Brazil vs Mexico": "round_of_16",
    "Belgium vs Japan": "round_of_16",
    "Sweden vs Switzerland": "round_of_16",
    "Colombia vs England": "round_of_16",

    # Cuartos de final
    "Uruguay vs France": "quarterfinals",
    "Brazil vs Belgium": "quarterfinals",
    "Sweden vs England": "quarterfinals",
    "Russia vs Croatia": "quarterfinals",

    # Semifinales
    "France vs Belgium": "semifinals",
    "Croatia vs England": "semifinals",

    # Tercer puesto
    "Belgium vs England (3rd Place)": "third_place",

    # Final
    "France vs Croatia": "final"
}


# =========================================================
# VALIDACIÓN
# =========================================================

missing_stage_matches = (
    set(MATCHES.keys())
    - set(MATCH_STAGES.keys())
)

unknown_stage_matches = (
    set(MATCH_STAGES.keys())
    - set(MATCHES.keys())
)

if missing_stage_matches:
    print(
        "❌ Partidos de MATCHES sin fase:",
        sorted(missing_stage_matches)
    )

if unknown_stage_matches:
    print(
        "❌ Partidos configurados que no existen en MATCHES:",
        sorted(unknown_stage_matches)
    )

if not missing_stage_matches and not unknown_stage_matches:
    print(
        "✅ Las fases de los",
        len(MATCH_STAGES),
        "partidos están correctamente configuradas."
    )
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 2. COBERTURA TEMPORAL DEL DATASET
# =========================================================

def get_match_coverage(
    df,
    match_name,
    interval_minutes=5
):
    """
    Calcula qué proporción de los intervalos del partido
    tiene registros disponibles en el dataset.
    """

    momentum, metadata = build_twitter_momentum_final(
        df=df,
        match_name=match_name,
        interval_minutes=interval_minutes
    )

    total_intervals = len(momentum)

    covered_intervals = int(
        momentum["has_coverage"].sum()
    )

    if total_intervals == 0:
        coverage_ratio = 0.0
    else:
        coverage_ratio = (
            covered_intervals
            / total_intervals
        )

    coverage_percentage = round(
        coverage_ratio * 100,
        1
    )

    if covered_intervals == 0:
        coverage_key = "none"
        coverage_label = "Sin cobertura temporal"

    elif coverage_ratio >= 0.80:
        coverage_key = "high"
        coverage_label = "Cobertura temporal alta"

    else:
        coverage_key = "partial"
        coverage_label = "Cobertura temporal parcial"

    return {
        "status": coverage_key,
        "label": coverage_label,
        "percentage": coverage_percentage,
        "covered_intervals": covered_intervals,
        "total_intervals": total_intervals
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 3. HERO DEL PARTIDO
# =========================================================

import re


SPANISH_MONTHS = {
    1: "enero",
    2: "febrero",
    3: "marzo",
    4: "abril",
    5: "mayo",
    6: "junio",
    7: "julio",
    8: "agosto",
    9: "septiembre",
    10: "octubre",
    11: "noviembre",
    12: "diciembre"
}


def format_match_date_spanish(date_value):
    """
    Convierte una fecha a un texto legible en español.
    """

    date = pd.to_datetime(date_value)

    return (
        f"{date.day} de "
        f"{SPANISH_MONTHS[date.month]} de "
        f"{date.year}"
    )


def extract_match_score(result_text):
    """
    Extrae el resultado principal de textos como:
    'France 4-3 Argentina'
    o
    'Spain 1-1 Russia (Russia gana por penales)'.
    """

    score_match = re.search(
        r"(\d+)\s*-\s*(\d+)",
        str(result_text)
    )

    if score_match is None:
        return {
            "team1": None,
            "team2": None,
            "display": ""
        }

    team1_score = int(score_match.group(1))
    team2_score = int(score_match.group(2))

    return {
        "team1": team1_score,
        "team2": team2_score,
        "display": f"{team1_score} - {team2_score}"
    }


def build_match_hero(
    df,
    match_name,
    interval_minutes=5
):
    """
    Construye la información necesaria para el Hero
    de cualquier partido.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    if match_name not in MATCH_EVENTS:
        raise ValueError(
            f"No existen eventos para: {match_name}"
        )

    if match_name not in MATCH_STAGES:
        raise ValueError(
            f"No existe fase configurada para: {match_name}"
        )

    match = MATCHES[match_name]
    historical_data = MATCH_EVENTS[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    stage_key = MATCH_STAGES[match_name]
    stage_label = MATCH_STAGE_LABELS[stage_key]

    result_text = historical_data["result"]
    winner = historical_data.get("winner")

    score = extract_match_score(
        result_text
    )

    raw_tweets = get_match_tweets(
        df,
        match_name
    )

    clean_tweets = clean_match_tweets(
        raw_tweets
    )

    coverage = get_match_coverage(
        df=df,
        match_name=match_name,
        interval_minutes=interval_minutes
    )

    period_layout = build_period_layout(
        match_name
    )

    period_keys = {
        period["key"]
        for period in period_layout
    }

    has_extra_time = (
        "extra_time_first_half"
        in period_keys
    )

    has_shootout = (
        "shootout"
        in period_keys
    )

    return {
        "match": match_name,

        "stage": {
            "key": stage_key,
            "label": stage_label
        },

        "date": {
            "iso": str(match["date"]),
            "display": format_match_date_spanish(
                match["date"]
            )
        },

        "team1": {
            "name": team1,
            "score": score["team1"],
            "flag_key": team1
        },

        "team2": {
            "name": team2,
            "score": score["team2"],
            "flag_key": team2
        },

        "score": score["display"],
        "result_text": result_text,
        "winner": winner,

        # Utilizamos los tweets limpios como cantidad
        # realmente analizada.
        "tweet_count": int(
            len(clean_tweets)
        ),

        "raw_tweet_count": int(
            len(raw_tweets)
        ),

        "coverage": coverage,

        "has_extra_time": has_extra_time,
        "has_shootout": has_shootout
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 4. SHARE OF VOICE
# =========================================================

def build_share_of_voice_final(
    df,
    match_name
):
    """
    Calcula qué proporción de las menciones corresponde
    a cada equipo.

    Un tuit que menciona a ambos equipos aporta una mención
    a cada uno.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    match = MATCHES[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    raw_tweets = get_match_tweets(
        df,
        match_name
    )

    clean_tweets = clean_match_tweets(
        raw_tweets
    )

    if clean_tweets.empty:
        return {
            "available": False,
            "match": match_name,
            "team1": {
                "name": team1,
                "mentions": 0,
                "percentage": 0.0
            },
            "team2": {
                "name": team2,
                "mentions": 0,
                "percentage": 0.0
            },
            "total_mentions": 0,
            "dominant_team": None,
            "difference": 0.0,
            "interpretation": (
                "No existe cobertura suficiente para "
                "calcular el dominio de la conversación."
            )
        }

    team1_pattern = build_keyword_pattern(
        match["keywords1"]
    )

    team2_pattern = build_keyword_pattern(
        match["keywords2"]
    )

    tweet_text = (
        clean_tweets["Tweet"]
        .fillna("")
        .astype(str)
    )

    team1_mentions = int(
        tweet_text.str.contains(
            team1_pattern,
            case=False,
            regex=True,
            na=False
        ).sum()
    )

    team2_mentions = int(
        tweet_text.str.contains(
            team2_pattern,
            case=False,
            regex=True,
            na=False
        ).sum()
    )

    total_mentions = (
        team1_mentions
        + team2_mentions
    )

    if total_mentions == 0:
        return {
            "available": False,
            "match": match_name,
            "team1": {
                "name": team1,
                "mentions": 0,
                "percentage": 0.0
            },
            "team2": {
                "name": team2,
                "mentions": 0,
                "percentage": 0.0
            },
            "total_mentions": 0,
            "dominant_team": None,
            "difference": 0.0,
            "interpretation": (
                "No se detectaron menciones suficientes "
                "para comparar a los equipos."
            )
        }

    team1_percentage = round(
        team1_mentions
        / total_mentions
        * 100,
        1
    )

    # Se calcula por diferencia para que la suma visible
    # sea siempre exactamente 100%.
    team2_percentage = round(
        100 - team1_percentage,
        1
    )

    difference = round(
        abs(
            team1_percentage
            - team2_percentage
        ),
        1
    )

    if team1_percentage > team2_percentage:
        dominant_team = team1
        dominant_percentage = team1_percentage
        other_team = team2

    elif team2_percentage > team1_percentage:
        dominant_team = team2
        dominant_percentage = team2_percentage
        other_team = team1

    else:
        dominant_team = None
        dominant_percentage = team1_percentage
        other_team = None

    if dominant_team is None:
        interpretation = (
            f"{team1} y {team2} tuvieron el mismo peso "
            f"en la conversación, con "
            f"{dominant_percentage:.1f}% cada uno."
        )

    else:
        interpretation = (
            f"{dominant_team} dominó la conversación "
            f"con el {dominant_percentage:.1f}% de las "
            f"menciones, una diferencia de "
            f"{difference:.1f} puntos porcentuales "
            f"respecto de {other_team}."
        )

    return {
        "available": True,
        "match": match_name,

        "team1": {
            "name": team1,
            "mentions": team1_mentions,
            "percentage": team1_percentage
        },

        "team2": {
            "name": team2,
            "mentions": team2_mentions,
            "percentage": team2_percentage
        },

        "total_mentions": total_mentions,
        "dominant_team": dominant_team,
        "difference": difference,
        "interpretation": interpretation
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 5. MOTOR DE EMOCIONES FUTBOLERAS
# =========================================================

import unicodedata


FOOTBALL_EMOTION_KEYWORDS = {
    "Euforia": [
        "goal",
        "gol",
        "goool",
        "goaaal",
        "win",
        "wins",
        "won",
        "winner",
        "winning",
        "victory",
        "champion",
        "champions",
        "celebrate",
        "celebration",
        "amazing",
        "incredible",
        "brilliant",
        "brilliance",
        "masterclass",
        "unreal",
        "stunning",
        "wonder goal",
        "wondergoal",
        "world class",
        "beautiful",
        "proud",
        "happy",
        "love",
        "yes",
        "yess",
        "vamos",
        "allez",
        "congratulations",
        "congrats",
        "parabens"
    ],

    "Tensión": [
        "nervous",
        "nerves",
        "tense",
        "tension",
        "penalty",
        "penalties",
        "shootout",
        "extra time",
        "overtime",
        "please",
        "hope",
        "scared",
        "fear",
        "worried",
        "suffering",
        "sufriendo",
        "nervios",
        "miedo",
        "por favor",
        "dios",
        "omg",
        "close game",
        "draw",
        "equalizer",
        "aguante"
    ],

    "Conflicto": [
        "referee",
        "ref",
        "var",
        "robbed",
        "robbery",
        "cheat",
        "cheating",
        "foul",
        "dive",
        "diving",
        "handball",
        "red card",
        "yellow card",
        "disgrace",
        "shame",
        "angry",
        "hate",
        "idiot",
        "stupid",
        "fuck",
        "shit",
        "arbitro",
        "robo",
        "robado",
        "mano",
        "falta",
        "tarjeta",
        "verguenza",
        "odio"
    ],

    "Tristeza": [
        "sad",
        "cry",
        "crying",
        "tears",
        "lost",
        "lose",
        "losing",
        "defeat",
        "defeated",
        "eliminated",
        "elimination",
        "heartbreak",
        "heartbroken",
        "devastated",
        "disappointed",
        "pain",
        "goodbye",
        "unlucky",
        "sorry",
        "perdimos",
        "derrota",
        "eliminado",
        "eliminados",
        "afuera",
        "fuera del mundial",
        "llorar",
        "llanto",
        "dolor",
        "triste",
        "chau",
        "adios",
        "lastima"
    ]
}


FOOTBALL_EMOTION_EMOJIS = {
    "Euforia": [
        "⚽", "🏆", "🎉", "🔥",
        "❤️", "👏", "😍", "🙌",
        "😀", "😁"
    ],

    "Tensión": [
        "😰", "😱", "🙏", "🤞", "😬"
    ],

    "Conflicto": [
        "😡", "🤬", "👎", "💢"
    ],

    "Tristeza": [
        "😭", "😢", "💔", "😞", "☹️"
    ]
}


EMOTION_TIE_PRIORITY = [
    "Conflicto",
    "Tristeza",
    "Tensión",
    "Euforia"
]


def normalize_emotion_text(text):
    """
    Normaliza el texto conservando emojis.
    """

    normalized = str(text).lower()

    normalized = unicodedata.normalize(
        "NFKD",
        normalized
    )

    normalized = "".join(
        character
        for character in normalized
        if not unicodedata.combining(character)
    )

    return normalized


def contains_emotion_term(
    normalized_text,
    term
):
    """
    Busca palabras o expresiones completas para evitar
    coincidencias parciales incorrectas.
    """

    pattern = (
        rf"(?<!\w)"
        rf"{re.escape(term)}"
        rf"(?!\w)"
    )

    return bool(
        re.search(
            pattern,
            normalized_text
        )
    )


def classify_football_emotion(
    text,
    return_scores=False
):
    """
    Clasifica un tuit como:
    Euforia, Tensión, Conflicto, Tristeza o Neutral.
    """

    normalized_text = normalize_emotion_text(
        text
    )

    scores = {
        emotion: 0
        for emotion in FOOTBALL_EMOTION_KEYWORDS
    }

    matched_terms = {
        emotion: []
        for emotion in FOOTBALL_EMOTION_KEYWORDS
    }

    for emotion, terms in (
        FOOTBALL_EMOTION_KEYWORDS.items()
    ):

        for term in terms:

            if contains_emotion_term(
                normalized_text,
                term
            ):
                scores[emotion] += 1
                matched_terms[emotion].append(
                    term
                )

    original_text = str(text)

    for emotion, emojis in (
        FOOTBALL_EMOTION_EMOJIS.items()
    ):

        for emoji in emojis:

            if emoji in original_text:
                scores[emotion] += 1
                matched_terms[emotion].append(
                    emoji
                )

    # La repetición de signos refuerza una emoción
    # solamente si ya existe una señal semántica.
    if (
        scores["Euforia"] > 0
        and original_text.count("!") >= 2
    ):
        scores["Euforia"] += 1

    if (
        scores["Tensión"] > 0
        and original_text.count("?") >= 2
    ):
        scores["Tensión"] += 1

    maximum_score = max(
        scores.values()
    )

    if maximum_score == 0:
        emotion = "Neutral"

    else:
        tied_emotions = [
            emotion_name
            for emotion_name, score
            in scores.items()
            if score == maximum_score
        ]

        emotion = next(
            priority_emotion
            for priority_emotion
            in EMOTION_TIE_PRIORITY
            if priority_emotion
            in tied_emotions
        )

    if return_scores:

        return {
            "emotion": emotion,
            "scores": scores,
            "matched_terms": matched_terms
        }

    return emotion
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 6. DISTRIBUCIÓN EMOCIONAL
# =========================================================

EMOTION_ORDER = [
    "Euforia",
    "Tensión",
    "Conflicto",
    "Tristeza"
]


EMOTION_COLORS = {
    "Euforia": "#2ECC71",
    "Tensión": "#F1C40F",
    "Conflicto": "#E67E22",
    "Tristeza": "#3498DB"
}


def calculate_emotion_summary(
    emotion_series
):
    """
    Resume una serie de emociones.

    Los porcentajes principales se calculan únicamente
    entre los tuits con evidencia emocional.
    """

    emotions = (
        pd.Series(emotion_series)
        .fillna("Neutral")
    )

    total_tweets = int(
        len(emotions)
    )

    counts = {
        emotion: int(
            (emotions == emotion).sum()
        )
        for emotion in EMOTION_ORDER
    }

    neutral_count = int(
        (emotions == "Neutral").sum()
    )

    classified_count = int(
        sum(counts.values())
    )

    if total_tweets == 0:
        classification_rate = 0.0
    else:
        classification_rate = round(
            classified_count
            / total_tweets
            * 100,
            1
        )

    if classified_count == 0:
        percentages = {
            emotion: 0.0
            for emotion in EMOTION_ORDER
        }

        predominant_emotion = None

    else:
        percentages = {
            emotion: round(
                counts[emotion]
                / classified_count
                * 100,
                1
            )
            for emotion in EMOTION_ORDER
        }

        # Ajuste para que la suma visible sea exactamente 100%.
        rounding_difference = round(
            100
            - sum(percentages.values()),
            1
        )

        predominant_emotion = max(
            EMOTION_ORDER,
            key=lambda emotion: counts[emotion]
        )

        percentages[predominant_emotion] = round(
            percentages[predominant_emotion]
            + rounding_difference,
            1
        )

    return {
        "total_tweets": total_tweets,
        "classified_tweets": classified_count,
        "neutral_tweets": neutral_count,
        "classification_rate": classification_rate,
        "counts": counts,
        "percentages": percentages,
        "predominant_emotion": predominant_emotion
    }


def build_emotional_climate_final(
    df,
    match_name,
    max_tweets=15000
):
    """
    Construye el clima emocional general y la comparación
    entre las conversaciones de ambos equipos.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    match = MATCHES[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    raw_tweets = get_match_tweets(
        df,
        match_name
    )

    clean_tweets = clean_match_tweets(
        raw_tweets
    )

    if clean_tweets.empty:
        return {
            "available": False,
            "match": match_name,
            "team1": {
                "name": team1,
                "summary":
                    calculate_emotion_summary([])
            },
            "team2": {
                "name": team2,
                "summary":
                    calculate_emotion_summary([])
            },
            "overall":
                calculate_emotion_summary([]),
            "sampled": False,
            "sample_size": 0,
            "interpretation": (
                "No existe cobertura suficiente para "
                "analizar el clima emocional."
            )
        }

    analysis_tweets = clean_tweets.copy()

    sampled = False

    if len(analysis_tweets) > max_tweets:

        analysis_tweets = (
            analysis_tweets
            .sample(
                n=max_tweets,
                random_state=42
            )
            .copy()
        )

        sampled = True

    analysis_tweets["_emotion"] = (
        analysis_tweets["Tweet"]
        .fillna("")
        .astype(str)
        .apply(
            classify_football_emotion
        )
    )

    team1_pattern = build_keyword_pattern(
        match["keywords1"]
    )

    team2_pattern = build_keyword_pattern(
        match["keywords2"]
    )

    tweet_text = (
        analysis_tweets["Tweet"]
        .fillna("")
        .astype(str)
    )

    team1_mask = tweet_text.str.contains(
        team1_pattern,
        case=False,
        regex=True,
        na=False
    )

    team2_mask = tweet_text.str.contains(
        team2_pattern,
        case=False,
        regex=True,
        na=False
    )

    overall_summary = (
        calculate_emotion_summary(
            analysis_tweets["_emotion"]
        )
    )

    team1_summary = (
        calculate_emotion_summary(
            analysis_tweets.loc[
                team1_mask,
                "_emotion"
            ]
        )
    )

    team2_summary = (
        calculate_emotion_summary(
            analysis_tweets.loc[
                team2_mask,
                "_emotion"
            ]
        )
    )

    predominant_emotion = (
        overall_summary[
            "predominant_emotion"
        ]
    )

    if predominant_emotion is None:

        interpretation = (
            "No se detectaron suficientes señales "
            "emocionales para determinar el clima "
            "predominante."
        )

    else:

        predominant_percentage = (
            overall_summary[
                "percentages"
            ][predominant_emotion]
        )

        interpretation = (
            f"{predominant_emotion} fue la emoción "
            f"predominante del partido, con el "
            f"{predominant_percentage:.1f}% de los "
            f"tuits que contenían señales emocionales."
        )

    return {
        "available": True,
        "match": match_name,

        "overall": overall_summary,

        "team1": {
            "name": team1,
            "summary": team1_summary
        },

        "team2": {
            "name": team2,
            "summary": team2_summary
        },

        "sampled": sampled,
        "sample_size": int(
            len(analysis_tweets)
        ),

        "interpretation": interpretation
    }
```

```python
# =========================================================
# VISUALIZACIÓN FINAL
# CLIMA EMOCIONAL
# =========================================================

def plot_emotional_climate_final(
    df,
    match_name
):
    """
    Muestra el clima emocional general y las distribuciones
    asociadas a las menciones de ambos equipos.
    """

    emotional_data = (
        build_emotional_climate_final(
            df=df,
            match_name=match_name
        )
    )

    team1 = emotional_data["team1"]["name"]
    team2 = emotional_data["team2"]["name"]

    figure = go.Figure()

    if not emotional_data["available"]:

        figure.add_annotation(
            x=0.5,
            y=0.5,
            xref="paper",
            yref="paper",
            text=(
                "NO HAY DATOS SUFICIENTES "
                "PARA ANALIZAR EL CLIMA EMOCIONAL"
            ),
            showarrow=False,
            font={
                "size": 17,
                "color": "#FFFFFF"
            }
        )

    else:

        row_positions = [2, 1, 0]

        for emotion in EMOTION_ORDER:

            percentages = [
                emotional_data[
                    "overall"
                ]["percentages"][emotion],

                emotional_data[
                    "team1"
                ]["summary"][
                    "percentages"
                ][emotion],

                emotional_data[
                    "team2"
                ]["summary"][
                    "percentages"
                ][emotion]
            ]

            counts = [
                emotional_data[
                    "overall"
                ]["counts"][emotion],

                emotional_data[
                    "team1"
                ]["summary"][
                    "counts"
                ][emotion],

                emotional_data[
                    "team2"
                ]["summary"][
                    "counts"
                ][emotion]
            ]

            visible_text = [
                (
                    f"<b>{percentage:.1f}%</b>"
                    if percentage >= 8
                    else ""
                )
                for percentage in percentages
            ]

            emotion_color = (
                EMOTION_COLORS[emotion]
            )

            figure.add_trace(
                go.Bar(
                    x=percentages,
                    y=row_positions,
                    orientation="h",
                    name=emotion,
                    width=0.48,

                    marker={
                        "color": emotion_color,
                        "line": {
                            "color":
                                "rgba(255,255,255,0.75)",
                            "width": 0.8
                        }
                    },

                    text=visible_text,
                    textposition="inside",
                    insidetextanchor="middle",

                    textfont={
                        "size": 13,
                        "color":
                            get_contrast_text_color(
                                emotion_color
                            )
                    },

                    customdata=counts,

                    hovertemplate=(
                        f"<b>{emotion}</b><br>"
                        "%{x:.1f}%<br>"
                        "%{customdata:,} tuits"
                        "<extra></extra>"
                    )
                )
            )

    # =====================================================
    # ETIQUETAS DE FILAS
    # =====================================================

    figure.add_annotation(
        x=0.13,
        y=2,
        xref="paper",
        yref="y",
        text="<b>PARTIDO</b>",
        showarrow=False,
        xanchor="center",
        font={
            "size": 13,
            "color": "#FFFFFF"
        }
    )

    figure.add_annotation(
        x=0.13,
        y=1,
        xref="paper",
        yref="y",
        text=f"<b>{team1.upper()}</b>",
        showarrow=False,
        xanchor="center",
        font={
            "size": 12,
            "color": "#FFFFFF"
        }
    )

    figure.add_annotation(
        x=0.13,
        y=0,
        xref="paper",
        yref="y",
        text=f"<b>{team2.upper()}</b>",
        showarrow=False,
        xanchor="center",
        font={
            "size": 12,
            "color": "#FFFFFF"
        }
    )

    # =====================================================
    # BANDERAS
    # =====================================================

    team1_flag = load_flag_data_uri(
        team1
    )

    team2_flag = load_flag_data_uri(
        team2
    )

    if team1_flag:

        figure.add_layout_image({
            "source": team1_flag,
            "xref": "paper",
            "yref": "y",
            "x": 0.035,
            "y": 1,
            "sizex": 0.065,
            "sizey": 0.52,
            "xanchor": "center",
            "yanchor": "middle",
            "sizing": "contain",
            "layer": "above"
        })

    if team2_flag:

        figure.add_layout_image({
            "source": team2_flag,
            "xref": "paper",
            "yref": "y",
            "x": 0.035,
            "y": 0,
            "sizex": 0.065,
            "sizey": 0.52,
            "xanchor": "center",
            "yanchor": "middle",
            "sizing": "contain",
            "layer": "above"
        })
    # =====================================================
    # LEYENDA EMOCIONAL PERSONALIZADA
    # =====================================================

    emotion_legend_positions = {
        "Euforia": 0.34,
        "Tensión": 0.49,
        "Conflicto": 0.65,
        "Tristeza": 0.82
    }

    for emotion, position in (
        emotion_legend_positions.items()
    ):

        emotion_color = (
            EMOTION_COLORS[emotion]
        )

        figure.add_annotation(
            x=position,
            y=-0.13,
            xref="paper",
            yref="paper",
            text=f"<b>{emotion}</b>",
            showarrow=False,

            bgcolor=emotion_color,

            bordercolor=emotion_color,
            borderwidth=1,
            borderpad=5,

            font={
                "size": 11,
                "color":
                    get_contrast_text_color(
                        emotion_color
                    )
            }
        )
    figure.update_layout(
        title={
            "text": (
                "<b>CLIMA EMOCIONAL</b>"
            ),
            "x": 0.5,
            "xanchor": "center",
            "y": 0.97,
            "font": {
                "family":
                    "Source Sans Pro, Arial, sans-serif",
                "size": 21,
                "color": "#FFFFFF"
            }
        },

        barmode="stack",
        height=340,

        margin={
            "l": 20,
            "r": 20,
            "t": 60,
            "b": 80
        },

        paper_bgcolor="rgba(0,0,0,0)",
        plot_bgcolor="rgba(0,0,0,0)",

        font={
            "family":
                "Source Sans Pro, Arial, sans-serif",
            "color": "#FFFFFF"
        },

        showlegend=False,

        hovermode="closest",

        xaxis={
            "domain": [0.22, 0.98],
            "range": [0, 100],
            "showgrid": False,
            "showticklabels": False,
            "showline": False,
            "zeroline": False,
            "fixedrange": True
        },

        yaxis={
            "range": [-0.6, 2.6],
            "showgrid": False,
            "showticklabels": False,
            "showline": False,
            "zeroline": False,
            "fixedrange": True
        },

        uniformtext={
            "minsize": 10,
            "mode": "hide"
        }
    )

    return figure, emotional_data
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 8. RANKING DE PROTAGONISTAS
# =========================================================

def build_player_pattern(
    aliases
):
    """
    Construye un patrón seguro para las variantes
    del nombre de un jugador.
    """

    normalized_aliases = [
        normalize_emotion_text(alias).strip()
        for alias in aliases
        if str(alias).strip()
    ]

    normalized_aliases = sorted(
        set(normalized_aliases),
        key=len,
        reverse=True
    )

    escaped_aliases = [
        re.escape(alias)
        for alias in normalized_aliases
    ]

    return (
        rf"(?<!\w)"
        rf"(?:{'|'.join(escaped_aliases)})"
        rf"(?!\w)"
    )


def build_protagonists_final(
    df,
    match_name,
    top_n=5
):
    """
    Encuentra los jugadores más mencionados de un partido.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    match = MATCHES[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    raw_tweets = get_match_tweets(
        df,
        match_name
    )

    clean_tweets = clean_match_tweets(
        raw_tweets
    )

    if clean_tweets.empty:
        return {
            "available": False,
            "match": match_name,
            "ranking": [],
            "total_player_mentions": 0,
            "top_player": None,
            "interpretation": (
                "No existe cobertura suficiente para "
                "identificar protagonistas."
            )
        }

    working_tweets = clean_tweets.copy()

    working_tweets["_normalized_text"] = (
        working_tweets["Tweet"]
        .fillna("")
        .astype(str)
        .apply(
            normalize_emotion_text
        )
    )

    likes = pd.to_numeric(
        working_tweets["Likes"],
        errors="coerce"
    ).fillna(0)

    retweets = pd.to_numeric(
        working_tweets["RTs"],
        errors="coerce"
    ).fillna(0)

    working_tweets["_engagement"] = (
        likes + retweets
    )

    player_rows = []

    for team in [team1, team2]:

        players = TEAM_PLAYER_ALIASES.get(
            team,
            {}
        )

        for player, aliases in players.items():

            player_pattern = (
                build_player_pattern(
                    aliases
                )
            )

            player_mask = (
                working_tweets[
                    "_normalized_text"
                ]
                .str.contains(
                    player_pattern,
                    regex=True,
                    na=False
                )
            )

            mentions = int(
                player_mask.sum()
            )

            engagement = int(
                working_tweets.loc[
                    player_mask,
                    "_engagement"
                ].sum()
            )

            player_rows.append({
                "player": player,
                "team": team,
                "mentions": mentions,
                "engagement": engagement
            })

    ranking = pd.DataFrame(
        player_rows
    )

    ranking = ranking[
        ranking["mentions"] > 0
    ].copy()

    if ranking.empty:
        return {
            "available": False,
            "match": match_name,
            "ranking": [],
            "total_player_mentions": 0,
            "top_player": None,
            "interpretation": (
                "No se detectaron menciones suficientes "
                "de los jugadores configurados."
            )
        }

    ranking = ranking.sort_values(
        by=[
            "mentions",
            "engagement",
            "player"
        ],
        ascending=[
            False,
            False,
            True
        ]
    ).reset_index(drop=True)

    total_player_mentions = int(
        ranking["mentions"].sum()
    )

    ranking["share"] = (
        ranking["mentions"]
        / total_player_mentions
        * 100
    ).round(1)

    top_ranking = (
        ranking
        .head(top_n)
        .copy()
    )

    top_ranking["rank"] = range(
        1,
        len(top_ranking) + 1
    )

    ranking_records = []

    for _, row in top_ranking.iterrows():

        ranking_records.append({
            "rank": int(row["rank"]),
            "player": row["player"],
            "team": row["team"],
            "mentions": int(
                row["mentions"]
            ),
            "share": float(
                row["share"]
            ),
            "engagement": int(
                row["engagement"]
            )
        })

    top_player = ranking_records[0]

    interpretation = (
        f"{top_player['player']} fue el jugador más "
        f"mencionado del partido, con "
        f"{top_player['mentions']:,} apariciones en "
        f"la conversación analizada."
    )

    return {
        "available": True,
        "match": match_name,
        "ranking": ranking_records,
        "total_player_mentions":
            total_player_mentions,
        "top_player": top_player,
        "interpretation": interpretation
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 9. TWEETS DESTACADOS
# =========================================================

def normalize_tweet_for_deduplication(
    text
):
    """
    Normaliza un tuit para detectar copias y duplicados.
    """

    normalized = normalize_emotion_text(
        text
    )

    normalized = re.sub(
        r"https?://\S+",
        " ",
        normalized
    )

    normalized = re.sub(
        r"\s+",
        " ",
        normalized
    )

    return normalized.strip()


def get_key_historical_event(
    match_name
):
    """
    Selecciona el evento con mayor importancia narrativa.
    En caso de empate conserva el primero.
    """

    events = MATCH_EVENTS[
        match_name
    ]["events"]

    if not events:
        return None

    return max(
        enumerate(events),
        key=lambda item: (
            item[1].get(
                "importance",
                1
            ),
            -item[0]
        )
    )[1]


def get_event_player_aliases(
    match_name,
    event
):
    """
    Busca qué jugador configurado aparece en la
    descripción del evento.
    """

    if event is None:
        return []

    match = MATCHES[match_name]

    event_text = normalize_emotion_text(
        event.get("event", "")
    )

    detected_aliases = []

    for team in [
        match["team1"],
        match["team2"]
    ]:

        for player, aliases in (
            TEAM_PLAYER_ALIASES[
                team
            ].items()
        ):

            normalized_player = (
                normalize_emotion_text(
                    player
                )
            )

            player_surname = (
                normalized_player
                .split()[-1]
            )

            normalized_aliases = [
                normalize_emotion_text(
                    alias
                )
                for alias in aliases
            ]

            player_detected = (
                player_surname in event_text
                or any(
                    alias in event_text
                    for alias
                    in normalized_aliases
                )
            )

            if player_detected:
                detected_aliases.extend(
                    aliases
                )

    return sorted(
        set(detected_aliases)
    )


def tweet_row_to_record(
    row,
    category_key,
    category_label,
    selection_method
):
    """
    Convierte una fila del DataFrame en información
    compatible con el futuro JSON.
    """

    tweet_date = row.get("Date")

    if pd.isna(tweet_date):
        date_iso = None
    else:
        date_iso = pd.to_datetime(
            tweet_date
        ).isoformat()

    return {
        "category": category_key,
        "label": category_label,
        "selection_method": selection_method,
        "text": str(
            row.get("Tweet", "")
        ),
        "date": date_iso,
        "likes": int(
            row.get("_likes", 0)
        ),
        "retweets": int(
            row.get("_retweets", 0)
        ),
        "engagement": int(
            row.get("_engagement", 0)
        ),
        "emotion": row.get(
            "_emotion",
            "Neutral"
        )
    }


def build_featured_tweets_final(
    df,
    match_name
):
    """
    Selecciona tres tuits distintos y relevantes,
    dentro del entorno temporal del partido.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    match = MATCHES[match_name]

    team1 = match["team1"]
    team2 = match["team2"]

    raw_tweets = get_match_tweets(
        df,
        match_name
    )

    clean_tweets = clean_match_tweets(
        raw_tweets
    )

    if clean_tweets.empty:
        return {
            "available": False,
            "match": match_name,
            "tweets": [],
            "key_event": None,
            "time_window_used": False
        }

    tweets = clean_tweets.copy()

    tweets["Date"] = pd.to_datetime(
        tweets["Date"],
        errors="coerce"
    )

    tweets = tweets.dropna(
        subset=["Date", "Tweet"]
    ).copy()

    # =====================================================
    # VENTANA TEMPORAL DEL PARTIDO
    # =====================================================

    kickoff = pd.to_datetime(
        MATCH_KICKOFFS[match_name]
    )

    runtime_periods = get_runtime_periods(
        match_name
    )

    runtime_minutes = sum(
        PERIOD_REAL_DURATIONS[
            period["key"]
        ]
        for period in runtime_periods
    )

    window_start = (
        kickoff
        - pd.Timedelta(minutes=30)
    )

    window_end = (
        kickoff
        + pd.Timedelta(
            minutes=runtime_minutes + 60
        )
    )

    match_window_tweets = tweets[
        tweets["Date"].between(
            window_start,
            window_end,
            inclusive="both"
        )
    ].copy()

    # Si hay registros en la ventana real, usamos solamente
    # esos registros. Si no, conservamos un fallback explícito.
    if len(match_window_tweets) >= 3:
        tweets = match_window_tweets
        time_window_used = True
    else:
        time_window_used = False

    # =====================================================
    # LIMPIEZA Y MÉTRICAS
    # =====================================================

    tweets = tweets[
        tweets["Tweet"]
        .fillna("")
        .astype(str)
        .str.strip()
        .ne("")
    ].copy()

    tweets["_dedup_text"] = (
        tweets["Tweet"]
        .apply(
            normalize_tweet_for_deduplication
        )
    )

    tweets = tweets[
        tweets["_dedup_text"].ne("")
    ].copy()

    tweets = tweets.drop_duplicates(
        subset=["_dedup_text"],
        keep="first"
    ).copy()

    tweets["_likes"] = pd.to_numeric(
        tweets["Likes"],
        errors="coerce"
    ).fillna(0)

    tweets["_retweets"] = pd.to_numeric(
        tweets["RTs"],
        errors="coerce"
    ).fillna(0)

    tweets["_engagement"] = (
        tweets["_likes"]
        + tweets["_retweets"]
    )

    tweets["_emotion"] = (
        tweets["Tweet"]
        .apply(
            classify_football_emotion
        )
    )

    # =====================================================
    # PUNTUACIÓN DE RELEVANCIA
    # =====================================================

    team1_pattern = build_player_pattern(
        [team1]
    )

    team2_pattern = build_player_pattern(
        [team2]
    )

    player_aliases = []

    for team in [team1, team2]:

        for aliases in (
            TEAM_PLAYER_ALIASES[
                team
            ].values()
        ):
            player_aliases.extend(
                aliases
            )

    player_pattern = build_player_pattern(
        player_aliases
    )

    tweets["_mentions_team1"] = (
        tweets["_dedup_text"]
        .str.contains(
            team1_pattern,
            regex=True,
            na=False
        )
    )

    tweets["_mentions_team2"] = (
        tweets["_dedup_text"]
        .str.contains(
            team2_pattern,
            regex=True,
            na=False
        )
    )

    tweets["_mentions_player"] = (
        tweets["_dedup_text"]
        .str.contains(
            player_pattern,
            regex=True,
            na=False
        )
    )

    tournament_teams = list(
        TEAM_PLAYER_ALIASES.keys()
    )

    def count_country_mentions(text):

        return sum(
            contains_emotion_term(
                text,
                normalize_emotion_text(
                    country
                )
            )
            for country in tournament_teams
        )

    tweets["_country_count"] = (
        tweets["_dedup_text"]
        .apply(
            count_country_mentions
        )
    )

    tweets["_emotion_signal"] = (
        tweets["_emotion"]
        .ne("Neutral")
        .astype(int)
    )
    tweets["_word_count"] = (
        tweets["_dedup_text"]
        .str.split()
        .str.len()
    )

    # Consideramos específico del partido un tuit que:
    # - menciona a los dos equipos, o
    # - menciona a un jugador y contiene una emoción.
    tweets["_match_specific"] = (
        (
            tweets["_mentions_team1"]
            &
            tweets["_mentions_team2"]
        )
        |
        (
            tweets["_mentions_player"]
            &
            tweets["_emotion"]
            .ne("Neutral")
        )
    )
    tweets["_relevance_score"] = (
        tweets["_mentions_team1"].astype(int)
        + tweets["_mentions_team2"].astype(int)
        + 2
        * tweets["_mentions_player"].astype(int)
        + tweets["_emotion_signal"]
    )

    # Los tuits que enumeran muchos países suelen ser
    # publicaciones genéricas sobre el torneo.
    tweets.loc[
        tweets["_country_count"] >= 4,
        "_relevance_score"
    ] -= 4
    relevant_tweets = tweets[
        (
            tweets["_relevance_score"] >= 2
        )
        &
        (
            tweets["_match_specific"]
        )
        &
        (
            tweets["_word_count"] >= 5
        )
        &
        (
            tweets["_country_count"] < 4
        )
    ].copy()

    # Fallback controlado: seguimos exigiendo jugadores
    # o ambos equipos y descartamos textos neutrales.
    if len(relevant_tweets) < 3:

        relevant_tweets = tweets[
            (
                (
                    tweets["_mentions_team1"]
                    &
                    tweets["_mentions_team2"]
                )
                |
                tweets["_mentions_player"]
            )
            &
            (
                tweets["_emotion"]
                != "Neutral"
            )
            &
            (
                tweets["_word_count"] >= 5
            )
            &
            (
                tweets["_country_count"] < 4
            )
        ].copy()

    if relevant_tweets.empty:

        return {
            "available": False,
            "match": match_name,
            "tweets": [],
            "key_event": None,
            "time_window_used":
                time_window_used
        }

    relevant_tweets = (
        relevant_tweets
        .sort_values(
            by=[
                "_engagement",
                "_relevance_score",
                "_retweets",
                "_likes"
            ],
            ascending=[
                False,
                False,
                False,
                False
            ]
        )
        .copy()
    )

    selected_indices = set()

    # =====================================================
    # 1. TUIT MÁS VIRAL Y RELEVANTE
    # =====================================================

    viral_index = relevant_tweets.index[0]

    viral_record = tweet_row_to_record(
        row=relevant_tweets.loc[
            viral_index
        ],
        category_key="most_viral",
        category_label="Tweet más viral",
        selection_method=(
            "relevant_match_tweet_and_engagement"
        )
    )

    selected_indices.add(
        viral_index
    )

    # =====================================================
    # 2. EVENTO CLAVE
    # =====================================================

    key_event = get_key_historical_event(
        match_name
    )

    event_aliases = (
        get_event_player_aliases(
            match_name,
            key_event
        )
    )

    key_moment_record = None

    if event_aliases:

        event_pattern = build_player_pattern(
            event_aliases
        )

        event_candidates = relevant_tweets[
            (
                relevant_tweets[
                    "_dedup_text"
                ]
                .str.contains(
                    event_pattern,
                    regex=True,
                    na=False
                )
            )
            &
            (
                relevant_tweets["_emotion"]
                != "Neutral"
            )
            &
            (
                ~relevant_tweets.index.isin(
                    selected_indices
                )
            )
        ]

        if not event_candidates.empty:

            event_index = (
                event_candidates.index[0]
            )

            key_moment_record = (
                tweet_row_to_record(
                    row=relevant_tweets.loc[
                        event_index
                    ],
                    category_key="key_moment",
                    category_label=(
                        "Reacción al momento clave"
                    ),
                    selection_method=(
                        "key_event_player_and_engagement"
                    )
                )
            )

            selected_indices.add(
                event_index
            )

    # =====================================================
    # 3. EMOCIÓN REPRESENTATIVA
    # =====================================================

    emotional_candidates = relevant_tweets[
        (
            relevant_tweets["_emotion"]
            != "Neutral"
        )
        &
        (
            ~relevant_tweets.index.isin(
                selected_indices
            )
        )
    ].copy()

    representative_record = None

    if not emotional_candidates.empty:

        predominant_emotion = (
            emotional_candidates[
                "_emotion"
            ]
            .value_counts()
            .idxmax()
        )

        representative_candidates = (
            emotional_candidates[
                emotional_candidates[
                    "_emotion"
                ]
                == predominant_emotion
            ]
        )

        if not representative_candidates.empty:

            representative_index = (
                representative_candidates.index[0]
            )

            representative_record = (
                tweet_row_to_record(
                    row=relevant_tweets.loc[
                        representative_index
                    ],
                    category_key="representative",
                    category_label=(
                        "La emoción del partido"
                    ),
                    selection_method=(
                        "relevant_predominant_emotion"
                    )
                )
            )

            selected_indices.add(
                representative_index
            )

    # =====================================================
    # FALLBACKS SIN DUPLICAR
    # =====================================================

    remaining_candidates = relevant_tweets[
        ~relevant_tweets.index.isin(
            selected_indices
        )
    ]

    if (
        representative_record is None
        and not remaining_candidates.empty
    ):

        representative_index = (
            remaining_candidates.index[0]
        )

        representative_record = (
            tweet_row_to_record(
                row=relevant_tweets.loc[
                    representative_index
                ],
                category_key="representative",
                category_label=(
                    "Reacción representativa"
                ),
                selection_method=(
                    "relevant_engagement_fallback"
                )
            )
        )

        selected_indices.add(
            representative_index
        )

    remaining_candidates = relevant_tweets[
        ~relevant_tweets.index.isin(
            selected_indices
        )
    ]

    if (
        key_moment_record is None
        and not remaining_candidates.empty
    ):

        reaction_index = (
            remaining_candidates.index[0]
        )

        key_moment_record = (
            tweet_row_to_record(
                row=relevant_tweets.loc[
                    reaction_index
                ],
                category_key=(
                    "featured_reaction"
                ),
                category_label=(
                    "Reacción destacada"
                ),
                selection_method=(
                    "relevant_engagement_fallback"
                )
            )
        )

    selected_tweets = [
        record
        for record in [
            viral_record,
            representative_record,
            key_moment_record
        ]
        if record is not None
    ]

    key_event_record = None

    if key_event is not None:

        key_event_record = {
            "minute": int(
                key_event["minute"]
            ),
            "event": key_event[
                "event"
            ],
            "importance": int(
                key_event.get(
                    "importance",
                    1
                )
            )
        }

    return {
        "available": bool(
            selected_tweets
        ),
        "match": match_name,
        "tweets": selected_tweets[:3],
        "key_event": key_event_record,
        "time_window_used":
            time_window_used,
        "window_start":
            window_start.isoformat(),
        "window_end":
            window_end.isoformat()
    }
```

```python
# =========================================================
# CORRECCIONES EDITORIALES DE TWEETS DESTACADOS
# =========================================================

def make_editorial_tweet(
    tweet_type,
    label,
    text,
    likes,
    retweets,
    emotion="Neutral"
):
    return {
        "type": tweet_type,
        "label": label,
        "text": text,
        "likes": int(likes),
        "retweets": int(retweets),
        "engagement": int(likes) + int(retweets),
        "emotion": emotion,
        "selection_method": "editorial_override"
    }


FEATURED_TWEET_OVERRIDES = {

    # -----------------------------------------------------
    # FRANCIA VS ARGENTINA
    # -----------------------------------------------------
    "France vs Argentina": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "Pace power and cool in front of goal every attribute "
                "centre halves at this tournament are scared of. "
                "World class today from Mbappe"
            ),
            likes=387,
            retweets=48,
            emotion="Euforia"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Lionel Messi at his four tournaments. The only player "
                "to provide an assist at each of the last four World Cups. "
                "Probably the last time we will see this man at a World Cup. "
                "Gutted for him"
            ),
            likes=247,
            retweets=113,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            "GAME ONN Pavard fires in his first France goal in style",
            likes=0,
            retweets=33,
            emotion="Euforia"
        )
    ],

    # -----------------------------------------------------
    # URUGUAY VS PORTUGAL
    # -----------------------------------------------------
    "Uruguay vs Portugal": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "The Greats have departed at the same stage on the same day. "
                "Ronaldo will be remembered at this World Cup"
            ),
            likes=0,
            retweets=2121
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            "Awww Ronaldo so cute and fair play helping Cavani",
            likes=0,
            retweets=1766
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción destacada",
            (
                "Neither Cristiano Ronaldo nor Lionel Messi have scored "
                "a goal in a World Cup knockout match"
            ),
            likes=0,
            retweets=1718,
            emotion="Tensión"
        )
    ],

    # -----------------------------------------------------
    # ESPAÑA VS RUSIA
    # -----------------------------------------------------
    "Spain vs Russia": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "Russia through to the quarters and Sergio Ramos is crying. "
                "BEST WORLD CUP EVER"
            ),
            likes=0,
            retweets=1657,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            "The end of an era. Andres Iniesta confirms his retirement after the exit",
            likes=0,
            retweets=288,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            (
                "Igor Akinfeev was the hero for Russia as he pulled off "
                "two brilliant saves to beat Spain on penalties"
            ),
            likes=0,
            retweets=387,
            emotion="Tensión"
        )
    ],

    # -----------------------------------------------------
    # CROACIA VS DINAMARCA
    # -----------------------------------------------------
    "Croatia vs Denmark": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "Kasper Schmeichel vs Croatia: saves a penalty in extra time, "
                "saves two penalties in the shootout, still knocked out. "
                "Heartbreaking"
            ),
            likes=1137,
            retweets=372,
            emotion="Tensión"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Luka Modric demonstrated his character and nerves of steel "
                "by stepping up in the shootout after missing earlier"
            ),
            likes=650,
            retweets=161,
            emotion="Tensión"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            (
                "Ivan Rakitic scored the winning penalty as Croatia beat "
                "Denmark to advance to the World Cup quarter-finals"
            ),
            likes=261,
            retweets=98,
            emotion="Tensión"
        )
    ],

    # -----------------------------------------------------
    # SUECIA VS SUIZA
    # -----------------------------------------------------
    "Sweden vs Switzerland": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "A lot of people found out that Sweden and Switzerland "
                "are not the same country today"
            ),
            likes=0,
            retweets=107
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Sad way for Manuel Akanji's excellent tournament to end, "
                "with the Sweden winner deflecting in off him"
            ),
            likes=0,
            retweets=48,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            "Emil Forsberg scores to give Sweden the lead",
            likes=0,
            retweets=31,
            emotion="Euforia"
        )
    ],

    # -----------------------------------------------------
    # COLOMBIA VS INGLATERRA
    # -----------------------------------------------------
    "Colombia vs England": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            (
                "Jordan Pickford was the first England goalkeeper to save "
                "a penalty in a major tournament shootout since 1998"
            ),
            likes=0,
            retweets=998,
            emotion="Tensión"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Jordan Henderson consoles a devastated James Rodriguez "
                "after the penalty shootout"
            ),
            likes=0,
            retweets=420,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            (
                "GOAL. Carlos Sanchez fouls Harry Kane from a corner "
                "and the captain takes care of the penalty"
            ),
            likes=0,
            retweets=382,
            emotion="Tensión"
        )
    ],

    # -----------------------------------------------------
    # FRANCIA VS BÉLGICA
    # -----------------------------------------------------
    "France vs Belgium": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            "BREAKING: France through to the final following a semi-final win over Belgium",
            likes=0,
            retweets=564,
            emotion="Euforia"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Full time. Belgium miss out on the final, "
                "but it has been an amazing tournament"
            ),
            likes=0,
            retweets=191,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            (
                "FRANCE ARE IN THE FINAL. Umtiti's header is enough "
                "to give them the win over Belgium"
            ),
            likes=0,
            retweets=290,
            emotion="Euforia"
        )
    ],

    # -----------------------------------------------------
    # CROACIA VS INGLATERRA
    # -----------------------------------------------------
    "Croatia vs England": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            "Croatia beat England to reach the World Cup final",
            likes=0,
            retweets=7179,
            emotion="Euforia"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Heartbreak for England as the team is knocked out "
                "of the World Cup by Croatia in the semi-final"
            ),
            likes=0,
            retweets=749,
            emotion="Tristeza"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al momento clave",
            (
                "CROATIA ARE IN THE FINAL. An extra-time goal from Mandzukic "
                "breaks England hearts"
            ),
            likes=0,
            retweets=261,
            emotion="Tensión"
        )
    ],

    # -----------------------------------------------------
    # FRANCIA VS CROACIA
    # -----------------------------------------------------
    "France vs Croatia": [
        make_editorial_tweet(
            "most_viral",
            "🔥 Tweet más viral",
            "FIFA Young Player Award: Kylian Mbappe",
            likes=0,
            retweets=14832,
            emotion="Euforia"
        ),
        make_editorial_tweet(
            "representative",
            "💬 La emoción del partido",
            (
                "Luka Modric is the player of the tournament, "
                "and that is absolutely deserved"
            ),
            likes=0,
            retweets=866,
            emotion="Euforia"
        ),
        make_editorial_tweet(
            "key_moment",
            "⚽ Reacción al desenlace",
            (
                "France overpower Croatia to lift the trophy "
                "for the second time in their history"
            ),
            likes=378,
            retweets=142,
            emotion="Euforia"
        )
    ]
}


# No mostramos tweets destacados cuando la muestra no permite
# construir una selección editorial suficientemente sólida.
FEATURED_TWEETS_HIDDEN_MATCHES = {
    "Brazil vs Mexico",
    "Belgium vs Japan"
}


def apply_featured_tweet_overrides(match_name, automatic_result):
    """
    Sustituye la selección automática únicamente cuando existe una
    selección editorial aprobada.

    Mantiene el formato original de build_featured_tweets_final().
    """

    if match_name in FEATURED_TWEETS_HIDDEN_MATCHES:
        editorial_tweets = []
        editorial_status = "hidden_due_to_limited_sample"

    elif match_name in FEATURED_TWEET_OVERRIDES:
        editorial_tweets = FEATURED_TWEET_OVERRIDES[match_name]
        editorial_status = "curated"

    else:
        return automatic_result

    # Si la función automática devuelve un diccionario
    if isinstance(automatic_result, dict):
        result = automatic_result.copy()

        possible_keys = [
            "tweets",
            "featured_tweets",
            "items",
            "cards"
        ]

        destination_key = next(
            (key for key in possible_keys if key in result),
            "tweets"
        )

        result[destination_key] = editorial_tweets
        result["available"] = bool(editorial_tweets)
        result["editorial_status"] = editorial_status

        return result

    # Si devuelve directamente una lista
    return editorial_tweets


print("✅ Correcciones editoriales cargadas")
print("Partidos curados:", len(FEATURED_TWEET_OVERRIDES))
print("Secciones ocultas por muestra insuficiente:", len(FEATURED_TWEETS_HIDDEN_MATCHES))
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 10. NOMBRES DE EQUIPOS EN ESPAÑOL
# =========================================================

TEAM_DISPLAY_NAMES_ES = {
    "France": "Francia",
    "Argentina": "Argentina",
    "Uruguay": "Uruguay",
    "Portugal": "Portugal",
    "Spain": "España",
    "Russia": "Rusia",
    "Croatia": "Croacia",
    "Denmark": "Dinamarca",
    "Brazil": "Brasil",
    "Mexico": "México",
    "Belgium": "Bélgica",
    "Japan": "Japón",
    "Sweden": "Suecia",
    "Switzerland": "Suiza",
    "Colombia": "Colombia",
    "England": "Inglaterra"
}


def get_team_display_name(team):
    """
    Devuelve el nombre del equipo en español.
    """

    return TEAM_DISPLAY_NAMES_ES.get(
        team,
        team
    )


def format_number_spanish(number):
    """
    Convierte 5208 en 5.208.
    """

    return f"{int(number):,}".replace(
        ",",
        "."
    )


def format_percentage_spanish(number):
    """
    Convierte 61.2 en 61,2.
    """

    return (
        f"{float(number):.1f}"
        .replace(".", ",")
    )
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 11. HISTORIA DEL PARTIDO
# =========================================================

EMOTION_NARRATIVE_PHRASES = {
    "Euforia": (
        "La celebración, el asombro y la ilusión "
        "marcaron el tono dominante de la conversación."
    ),

    "Tensión": (
        "La conversación estuvo atravesada por la "
        "incertidumbre, los nervios y la sensación de "
        "que el partido podía cambiar en cualquier momento."
    ),

    "Conflicto": (
        "Las discusiones, las críticas y la polémica "
        "ocuparon el centro de la conversación."
    ),

    "Tristeza": (
        "La decepción y la tristeza fueron las señales "
        "emocionales más presentes en la conversación."
    )
}


def build_match_story_final(
    match_name,
    hero,
    share_of_voice,
    emotional_climate,
    protagonists,
    featured_tweets
):
    """
    Transforma los análisis del partido en una historia
    periodística dividida en cuatro bloques.
    """

    team1 = hero["team1"]["name"]
    team2 = hero["team2"]["name"]

    team1_es = get_team_display_name(
        team1
    )

    team2_es = get_team_display_name(
        team2
    )

    winner = hero["winner"]

    winner_es = (
        get_team_display_name(winner)
        if winner
        else None
    )

    stage_label = hero[
        "stage"
    ]["label"].lower()

    tweet_count = format_number_spanish(
        hero["tweet_count"]
    )

    # =====================================================
    # 1. LA PREVIA
    # =====================================================

    preview_text = (
        f"{team1_es} y {team2_es} se encontraron en "
        f"{stage_label} del Mundial de Rusia 2018. "
        f"El encuentro dejó una huella social formada por "
        f"{tweet_count} tuits limpios, una muestra de las "
        f"reacciones, expectativas y discusiones que "
        f"acompañaron al partido."
    )

    # =====================================================
    # 2. EL CLIMA SOCIAL
    # =====================================================

    social_parts = []

    if share_of_voice["available"]:

        dominant_team = (
            share_of_voice[
                "dominant_team"
            ]
        )

        dominant_team_es = (
            get_team_display_name(
                dominant_team
            )
        )

        if (
            dominant_team
            == team1
        ):
            dominant_percentage = (
                share_of_voice[
                    "team1"
                ]["percentage"]
            )
        else:
            dominant_percentage = (
                share_of_voice[
                    "team2"
                ]["percentage"]
            )

        dominant_percentage_text = (
            format_percentage_spanish(
                dominant_percentage
            )
        )

        social_parts.append(
            f"{dominant_team_es} concentró el "
            f"{dominant_percentage_text}% de las menciones "
            f"y se convirtió en el principal foco de "
            f"atención de Twitter."
        )

    predominant_emotion = (
        emotional_climate[
            "overall"
        ]["predominant_emotion"]
    )

    if predominant_emotion:

        emotion_percentage = (
            emotional_climate[
                "overall"
            ]["percentages"][
                predominant_emotion
            ]
        )

        emotion_percentage_text = (
            format_percentage_spanish(
                emotion_percentage
            )
        )

        social_parts.append(
            EMOTION_NARRATIVE_PHRASES[
                predominant_emotion
            ]
            + " "
            + f"{predominant_emotion} representó el "
            + f"{emotion_percentage_text}% de los tuits "
            + "con señales emocionales."
        )

    if (
        protagonists["available"]
        and protagonists["top_player"]
    ):

        top_player = (
            protagonists[
                "top_player"
            ]
        )

        player_mentions = (
            format_number_spanish(
                top_player["mentions"]
            )
        )

        social_parts.append(
            f"En ese escenario, {top_player['player']} "
            f"emergió como el protagonista más mencionado, "
            f"con {player_mentions} apariciones."
        )

    if social_parts:
        social_climate_text = " ".join(
            social_parts
        )
    else:
        social_climate_text = (
            "La cobertura disponible no permite definir "
            "con suficiente claridad qué equipo o emoción "
            "dominó la conversación."
        )

    # =====================================================
    # 3. EL PUNTO DE QUIEBRE
    # =====================================================

    key_event = featured_tweets.get(
        "key_event"
    )

    if key_event:

        key_minute = key_event[
            "minute"
        ]

        key_description = key_event[
            "event"
        ]

        turning_point_text = (
            f"El punto de quiebre de la reconstrucción "
            f"deportiva llegó en el minuto {key_minute}: "
            f"{key_description}. "
            f"Por su importancia y por el impacto que tuvo "
            f"sobre el desarrollo del encuentro, ese momento "
            f"resume el cambio central de la historia."
        )

    else:
        turning_point_text = (
            "La información disponible no permite "
            "identificar con suficiente precisión un único "
            "punto de quiebre."
        )

    # =====================================================
    # 4. EL DESENLACE
    # =====================================================

    score = hero["score"]

    if winner_es:

        outcome_start = (
            f"El partido terminó {score} y "
            f"{winner_es} se quedó con la victoria."
        )

    else:
        outcome_start = (
            f"El encuentro terminó {score}."
        )

    if hero["has_shootout"]:

        outcome_start += (
            " La clasificación se resolvió en la "
            "tanda de penales."
        )

    dominant_team = (
        share_of_voice.get(
            "dominant_team"
        )
    )

    if (
        winner
        and dominant_team
        and winner != dominant_team
    ):

        dominant_team_es = (
            get_team_display_name(
                dominant_team
            )
        )

        closing_sentence = (
            f"{dominant_team_es} dominó la conversación, "
            f"pero {winner_es} terminó escribiendo el "
            f"desenlace deportivo."
        )

    elif (
        winner
        and dominant_team
        and winner == dominant_team
    ):

        closing_sentence = (
            f"{winner_es} dominó tanto la atención de "
            f"Twitter como el desenlace del partido."
        )

    else:
        closing_sentence = (
            "La conversación y el resultado dejaron dos "
            "formas diferentes de recordar el encuentro."
        )

    outcome_text = (
        outcome_start
        + " "
        + closing_sentence
    )

    # =====================================================
    # NOTA DE COBERTURA
    # =====================================================

    coverage = hero[
        "coverage"
    ]

    coverage_percentage = (
        format_percentage_spanish(
            coverage["percentage"]
        )
    )

    if coverage["status"] == "high":

        coverage_note = (
            f"Cobertura temporal alta: el dataset contiene "
            f"registros en el {coverage_percentage}% de los "
            f"intervalos analizados."
        )

    elif coverage["status"] == "partial":

        coverage_note = (
            f"Cobertura temporal parcial: el dataset contiene "
            f"registros en el {coverage_percentage}% de los "
            f"intervalos. La historia evita atribuir reacciones "
            f"a minutos sin datos."
        )

    else:

        coverage_note = (
            "El dataset no contiene cobertura temporal del "
            "desarrollo del partido. La reconstrucción se "
            "apoya en la información histórica y en los "
            "registros sociales disponibles fuera de la "
            "línea temporal."
        )

    return {
        "match": match_name,

        "preview": {
            "title": "La previa",
            "text": preview_text
        },

        "social_climate": {
            "title": "El clima social",
            "text": social_climate_text
        },

        "turning_point": {
            "title": "El punto de quiebre",
            "text": turning_point_text
        },

        "outcome": {
            "title": "El desenlace",
            "text": outcome_text
        },

        "coverage_note": coverage_note
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 12. RADIOGRAFÍA FINAL
# =========================================================

def get_momentum_peak_summary(
    momentum
):
    """
    Encuentra el intervalo con mayor cantidad combinada
    de menciones de ambos equipos.
    """

    if momentum is None or momentum.empty:
        return None

    available = momentum[
        momentum["has_coverage"]
    ].copy()

    if available.empty:
        return None

    available["_combined_mentions"] = (
        available["team1_mentions"]
        .fillna(0)
        +
        available["team2_mentions"]
        .fillna(0)
    )

    if (
        available["_combined_mentions"]
        .max()
        <= 0
    ):
        return None

    peak_index = (
        available["_combined_mentions"]
        .idxmax()
    )

    peak_row = available.loc[
        peak_index
    ]

    period_label = peak_row.get(
        "period_label",
        "Intervalo de máxima actividad"
    )

    return {
        "label": str(period_label),
        "mentions": int(
            peak_row["_combined_mentions"]
        ),
        "display_minute": float(
            peak_row["minute"]
        )
    }


def build_final_radiography(
    match_name,
    hero,
    share_of_voice,
    emotional_climate,
    protagonists,
    featured_tweets,
    momentum
):
    """
    Construye el resumen ejecutivo final del partido.
    """

    team1 = hero["team1"]["name"]
    team2 = hero["team2"]["name"]

    team1_es = get_team_display_name(
        team1
    )

    team2_es = get_team_display_name(
        team2
    )

    winner = hero["winner"]

    winner_es = (
        get_team_display_name(
            winner
        )
        if winner
        else "Sin definir"
    )

    score_team1 = hero[
        "team1"
    ]["score"]

    score_team2 = hero[
        "team2"
    ]["score"]

    result_display = (
        f"{team1_es} "
        f"{score_team1}–{score_team2} "
        f"{team2_es}"
    )

    # =====================================================
    # DOMINIO DE LA CONVERSACIÓN
    # =====================================================

    dominant_team = (
        share_of_voice.get(
            "dominant_team"
        )
    )

    if dominant_team:

        dominant_team_es = (
            get_team_display_name(
                dominant_team
            )
        )

        if (
            dominant_team
            == team1
        ):
            dominant_percentage = (
                share_of_voice[
                    "team1"
                ]["percentage"]
            )
        else:
            dominant_percentage = (
                share_of_voice[
                    "team2"
                ]["percentage"]
            )

        conversation_value = (
            f"{dominant_team_es} "
            f"{format_percentage_spanish(dominant_percentage)}%"
        )

    else:
        dominant_team_es = None
        conversation_value = (
            "Sin datos suficientes"
        )

    # =====================================================
    # PROTAGONISTA
    # =====================================================

    top_player = protagonists.get(
        "top_player"
    )

    if top_player:

        protagonist_value = (
            top_player["player"]
        )

        protagonist_detail = (
            f"{format_number_spanish(top_player['mentions'])} "
            f"menciones"
        )

    else:
        protagonist_value = (
            "Sin datos suficientes"
        )

        protagonist_detail = ""

    # =====================================================
    # EMOCIÓN
    # =====================================================

    predominant_emotion = (
        emotional_climate[
            "overall"
        ]["predominant_emotion"]
    )

    if predominant_emotion:

        emotion_percentage = (
            emotional_climate[
                "overall"
            ]["percentages"][
                predominant_emotion
            ]
        )

        emotion_value = (
            predominant_emotion
        )

        emotion_detail = (
            f"{format_percentage_spanish(emotion_percentage)}% "
            f"de las señales emocionales"
        )

    else:
        emotion_value = (
            "Sin datos suficientes"
        )

        emotion_detail = ""

    # =====================================================
    # PICO DE ACTIVIDAD
    # =====================================================

    peak = get_momentum_peak_summary(
        momentum
    )

    if peak:

        peak_value = peak["label"]

        peak_detail = (
            f"{format_number_spanish(peak['mentions'])} "
            f"menciones combinadas"
        )

    else:
        peak_value = (
            "Sin cobertura temporal"
        )

        peak_detail = ""

    # =====================================================
    # TARJETAS
    # =====================================================

    cards = [
        {
            "key": "result",
            "label": "Resultado",
            "value": result_display,
            "detail": (
                f"Ganador: {winner_es}"
            )
        },

        {
            "key": "tweets",
            "label": "Tweets analizados",
            "value": format_number_spanish(
                hero["tweet_count"]
            ),
            "detail": (
                hero["coverage"]["label"]
            )
        },

        {
            "key": "conversation",
            "label": "Dominio de la conversación",
            "value": conversation_value,
            "detail": (
                f"Diferencia de "
                f"{format_percentage_spanish(share_of_voice['difference'])} "
                f"puntos"
                if share_of_voice["available"]
                else ""
            )
        },

        {
            "key": "protagonist",
            "label": "Protagonista",
            "value": protagonist_value,
            "detail": protagonist_detail
        },

        {
            "key": "emotion",
            "label": "Emoción predominante",
            "value": emotion_value,
            "detail": emotion_detail
        },

        {
            "key": "peak",
            "label": "Pico de actividad",
            "value": peak_value,
            "detail": peak_detail
        }
    ]

    # =====================================================
    # FRASE FINAL
    # =====================================================

    if (
        dominant_team_es
        and top_player
        and winner_es
    ):

        final_sentence = (
            f"{dominant_team_es} dominó la conversación; "
            f"{top_player['player']} fue el nombre propio "
            f"y {winner_es} se quedó con el resultado."
        )

    elif winner_es:

        final_sentence = (
            f"{winner_es} se quedó con el resultado de un "
            f"partido cuya conversación dejó múltiples "
            f"formas de recordarlo."
        )

    else:

        final_sentence = (
            "El partido dejó una historia deportiva y otra "
            "historia paralela escrita en Twitter."
        )

    return {
        "match": match_name,
        "cards": cards,
        "final_sentence": final_sentence
    }
```

```python
# =========================================================
# CALIDAD Y DISPONIBILIDAD DE CADA PARTIDO
# =========================================================

def classify_match_data_quality(
    hero
):
    """
    Determina qué experiencia puede ofrecer la aplicación
    según la cantidad de tuits y la cobertura temporal.
    """

    tweet_count = hero[
        "tweet_count"
    ]

    coverage_percentage = hero[
        "coverage"
    ]["percentage"]

    if tweet_count == 0:

        return {
            "level": "unavailable",
            "label": "Sin datos sociales",
            "message": (
                "El dataset no contiene tuits disponibles "
                "para la fecha de este partido."
            ),
            "show_social_analysis": False,
            "show_momentum": False,
            "show_featured_tweets": False
        }

    if coverage_percentage > 0:

        return {
            "level": "partial_timeline",
            "label": "Cobertura temporal parcial",
            "message": (
                "El dataset permite reconstruir solamente "
                "algunos tramos de la conversación."
            ),
            "show_social_analysis": True,
            "show_momentum": True,
            "show_featured_tweets": True
        }

    if tweet_count < 500:

        return {
            "level": "limited_text",
            "label": "Muestra social limitada",
            "message": (
                "Existen tuits relacionados con el partido, "
                "pero la muestra es pequeña y no cubre su "
                "desarrollo temporal."
            ),
            "show_social_analysis": True,
            "show_momentum": False,
            "show_featured_tweets": True
        }

    return {
        "level": "text_only",
        "label": "Análisis social sin línea temporal",
        "message": (
            "Existen suficientes tuits para analizar la "
            "conversación, pero no para reconstruir su "
            "evolución minuto a minuto."
        ),
        "show_social_analysis": True,
        "show_momentum": False,
        "show_featured_tweets": True
    }
```

```python
# =========================================================
# PAQUETE FINAL DEL PARTIDO
# 13. GENERADOR DEL PAQUETE COMPLETO
# =========================================================

import json
from datetime import datetime, date


MATCH_PACKAGE_VERSION = "2.0"


def slugify_match_name(
    match_name
):
    """
    Convierte el nombre del partido en un nombre
    seguro para archivos.
    """

    slug = normalize_emotion_text(
        match_name
    )

    slug = re.sub(
        r"[^a-z0-9]+",
        "-",
        slug
    )

    return slug.strip("-")


def make_json_safe(value):
    """
    Convierte valores de Pandas y NumPy a tipos
    compatibles con JSON.
    """

    if value is None:
        return None

    if isinstance(value, dict):
        return {
            str(key): make_json_safe(item)
            for key, item in value.items()
        }

    if isinstance(
        value,
        (list, tuple, set)
    ):
        return [
            make_json_safe(item)
            for item in value
        ]

    if isinstance(
        value,
        (pd.Timestamp, datetime, date)
    ):
        return value.isoformat()

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):

        if np.isnan(value):
            return None

        return float(value)

    if isinstance(value, np.bool_):
        return bool(value)

    if isinstance(value, float):

        if np.isnan(value):
            return None

        return value

    try:
        if pd.isna(value):
            return None
    except (TypeError, ValueError):
        pass

    return value


def generate_match_package(
    df,
    match_name,
    interval_minutes=5
):
    """
    Ejecuta todos los módulos y devuelve el paquete
    completo de un partido.
    """

    if match_name not in MATCHES:
        raise ValueError(
            f"Partido desconocido: {match_name}"
        )

    # =====================================================
    # ANÁLISIS PRINCIPALES
    # =====================================================

    hero = build_match_hero(
        df=df,
        match_name=match_name,
        interval_minutes=interval_minutes
    )
    quality = classify_match_data_quality(
        hero
    )
    momentum, momentum_metadata = (
        build_twitter_momentum_final(
            df=df,
            match_name=match_name,
            interval_minutes=interval_minutes
        )
    )

    share_of_voice = (
        build_share_of_voice_final(
            df=df,
            match_name=match_name
        )
    )

    emotional_climate = (
        build_emotional_climate_final(
            df=df,
            match_name=match_name
        )
    )

    protagonists = (
        build_protagonists_final(
            df=df,
            match_name=match_name,
            top_n=5
        )
    )

    featured_tweets = (
        build_featured_tweets_final(
            df=df,
            match_name=match_name
        ))
    featured_tweets = apply_featured_tweet_overrides(
    match_name,
    featured_tweets
)
    has_featured_tweets = bool(
        featured_tweets.get(
            "tweets",
            []
        )
    )

    quality["show_featured_tweets"] = (
        has_featured_tweets
    )

    if has_featured_tweets:

        featured_tweets_review = (
            featured_tweets.get(
                "editorial_status",
                "automatic"
            )
        )

    else:

        featured_tweets_review = (
            featured_tweets.get(
                "editorial_status",
                "not_available"
            )
        )
    # =====================================================
    # NARRATIVA Y CIERRE
    # =====================================================

    story = build_match_story_final(
        match_name=match_name,
        hero=hero,
        share_of_voice=share_of_voice,
        emotional_climate=emotional_climate,
        protagonists=protagonists,
        featured_tweets=featured_tweets
    )

    radiography = build_final_radiography(
        match_name=match_name,
        hero=hero,
        share_of_voice=share_of_voice,
        emotional_climate=emotional_climate,
        protagonists=protagonists,
        featured_tweets=featured_tweets,
        momentum=momentum
    )

    # =====================================================
    # SERIE DEL PULSO
    # =====================================================

    momentum_records = (
        momentum
        .to_dict(
            orient="records"
        )
    )

    package = {
        "schema_version":
            MATCH_PACKAGE_VERSION,

        "status": "ready",

        "slug":
            slugify_match_name(
                match_name
            ),

        "match":
            match_name,

        "hero":
            hero,

        "story":
            story,

        "momentum": {
            "interval_minutes":
                interval_minutes,

            "metadata":
                momentum_metadata,

            "series":
                momentum_records
        },

        "share_of_voice":
            share_of_voice,

        "emotional_climate":
            emotional_climate,

        "protagonists":
            protagonists,

        "featured_tweets":
            featured_tweets,

        "radiography":
            radiography,

        "data_quality": {
            **quality,

            "coverage":
                hero["coverage"],

            "featured_tweets_review":
                featured_tweets_review,

            "warning":
                story["coverage_note"]
        }
    }

    return make_json_safe(
        package
    )
```

```python
# =========================================================
# AUDITORÍA GENERAL
# GENERAR TODOS LOS PARTIDOS EN MEMORIA
# =========================================================

def generate_all_match_packages(
    df,
    interval_minutes=5
):
    """
    Genera todos los paquetes sin guardarlos todavía.
    También crea una tabla de auditoría.
    """

    packages = {}
    audit_rows = []

    match_names = list(
        MATCHES.keys()
    )

    total_matches = len(
        match_names
    )

    for position, match_name in enumerate(
        match_names,
        start=1
    ):

        print(
            f"[{position}/{total_matches}]",
            match_name
        )

        try:

            package = generate_match_package(
                df=df,
                match_name=match_name,
                interval_minutes=interval_minutes
            )

            packages[match_name] = package

            emotion_rate = (
                package[
                    "emotional_climate"
                ]["overall"][
                    "classification_rate"
                ]
            )

            audit_rows.append({
                "match": match_name,

                "status": "OK",

                "tweets": package[
                    "hero"
                ]["tweet_count"],

                "coverage_status": package[
                    "hero"
                ]["coverage"]["status"],

                "coverage_percentage": package[
                    "hero"
                ]["coverage"]["percentage"],

                "momentum_intervals": len(
                    package[
                        "momentum"
                    ]["series"]
                ),

                "emotion_rate": emotion_rate,

                "protagonists": len(
                    package[
                        "protagonists"
                    ]["ranking"]
                ),

                "featured_tweets": len(
                    package[
                        "featured_tweets"
                    ]["tweets"]
                ),

                "error": None
            })

        except Exception as error:

            audit_rows.append({
                "match": match_name,
                "status": "ERROR",
                "tweets": 0,
                "coverage_status": None,
                "coverage_percentage": 0.0,
                "momentum_intervals": 0,
                "emotion_rate": 0.0,
                "protagonists": 0,
                "featured_tweets": 0,
                "error": str(error)
            })

            print(
                "   ❌",
                type(error).__name__,
                ":",
                error
            )

    audit = pd.DataFrame(
        audit_rows
    )

    return packages, audit
```

```python
# =========================================================
# EJECUTAR AUDITORÍA DE LOS 16 PARTIDOS
# =========================================================

ALL_MATCH_PACKAGES, MATCH_AUDIT = (
    generate_all_match_packages(
        df=df,
        interval_minutes=5
    )
)

print(
    "\nPaquetes generados:",
    len(ALL_MATCH_PACKAGES),
    "/",
    len(MATCHES)
)

display(
    MATCH_AUDIT
)
```

```python
# =========================================================
# VALIDACIÓN FINAL DE LOS 16 PAQUETES
# =========================================================

REQUIRED_PACKAGE_SECTIONS = {
    "hero",
    "story",
    "momentum",
    "share_of_voice",
    "emotional_climate",
    "protagonists",
    "featured_tweets",
    "radiography",
    "data_quality"
}

validation_rows = []

for match_name, package in ALL_MATCH_PACKAGES.items():

    errors = []

    missing_sections = (
        REQUIRED_PACKAGE_SECTIONS
        - set(package.keys())
    )

    if missing_sections:
        errors.append(
            "Faltan secciones: "
            + ", ".join(
                sorted(missing_sections)
            )
        )

    try:
        json.dumps(
            package,
            ensure_ascii=False
        )

    except Exception as error:
        errors.append(
            f"JSON inválido: {error}"
        )

    featured_tweets = (
        package
        .get("featured_tweets", {})
        .get("tweets", [])
    )

    show_featured_tweets = (
        package
        .get("data_quality", {})
        .get("show_featured_tweets")
    )

    if (
        bool(featured_tweets)
        != bool(show_featured_tweets)
    ):
        errors.append(
            "Inconsistencia en Tweets destacados"
        )

    editorial_status = (
        package
        .get("featured_tweets", {})
        .get("editorial_status")
    )

    if editorial_status == "curated":

        untranslated = [
            tweet
            for tweet in featured_tweets
            if not tweet.get("translation_es")
        ]

        if untranslated:
            errors.append(
                "Faltan traducciones editoriales"
            )

    tweet_count = (
        package
        .get("hero", {})
        .get("tweet_count", 0)
    )

    show_social_analysis = (
        package
        .get("data_quality", {})
        .get("show_social_analysis")
    )

    if (
        tweet_count == 0
        and show_social_analysis
    ):
        errors.append(
            "Se muestra análisis social sin tuits"
        )

    validation_rows.append({
        "match": match_name,
        "validation": (
            "OK"
            if not errors
            else "ERROR"
        ),
        "details": (
            None
            if not errors
            else " | ".join(errors)
        )
    })


FINAL_PACKAGE_VALIDATION = pd.DataFrame(
    validation_rows
)

display(
    FINAL_PACKAGE_VALIDATION
)

assert (
    FINAL_PACKAGE_VALIDATION[
        "validation"
    ]
    == "OK"
).all(), (
    "Existen paquetes que requieren corrección"
)

print(
    "\n✅ Validación final superada:",
    len(FINAL_PACKAGE_VALIDATION),
    "/",
    len(MATCHES)
)
```

```python
# =========================================================
# EXPORTACIÓN FINAL PARA STREAMLIT
# =========================================================

import os
import json
from datetime import datetime


PROJECT_DRIVE_DIR = (
    "/content/drive/MyDrive/"
    "WorldCupStoryteller"
)

DATA_EXPORT_DIR = os.path.join(
    PROJECT_DRIVE_DIR,
    "data"
)

MATCHES_EXPORT_DIR = os.path.join(
    DATA_EXPORT_DIR,
    "matches"
)

os.makedirs(
    MATCHES_EXPORT_DIR,
    exist_ok=True
)


# =========================================================
# EXPORTAR LOS 16 PARTIDOS
# =========================================================

exported_files = []
index_matches = []

for match_name, package in (
    ALL_MATCH_PACKAGES.items()
):

    slug = package["slug"]

    file_name = f"{slug}.json"

    file_path = os.path.join(
        MATCHES_EXPORT_DIR,
        file_name
    )

    with open(
        file_path,
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            package,
            file,
            ensure_ascii=False,
            indent=2
        )

    exported_files.append(
        file_path
    )

    hero = package["hero"]

    index_matches.append({
        "match": match_name,
        "slug": slug,
        "status": package["status"],
        "stage": hero["stage"],
        "date": hero["date"],
        "team1": hero["team1"]["name"],
        "team2": hero["team2"]["name"],
        "score": hero["score"],
        "winner": hero["winner"],
        "data_quality": package[
            "data_quality"
        ]["level"],
        "file": (
            f"matches/{file_name}"
        )
    })


# =========================================================
# ÍNDICE PARA LA NAVEGACIÓN DE STREAMLIT
# =========================================================

match_index = {
    "schema_version":
        MATCH_PACKAGE_VERSION,

    "generated_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "match_count":
        len(index_matches),

    "matches":
        index_matches
}

index_file_path = os.path.join(
    DATA_EXPORT_DIR,
    "index.json"
)

with open(
    index_file_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        match_index,
        file,
        ensure_ascii=False,
        indent=2
    )


# =========================================================
# VALIDACIÓN DE LA EXPORTACIÓN
# =========================================================

assert len(exported_files) == 16, (
    "No se exportaron los 16 partidos"
)

assert os.path.exists(
    index_file_path
), (
    "No se creó index.json"
)

print(
    "✅ Exportación final completada"
)

print(
    "Partidos exportados:",
    len(exported_files)
)

print(
    "Índice:",
    index_file_path
)

print(
    "Carpeta:",
    MATCHES_EXPORT_DIR
)
```
